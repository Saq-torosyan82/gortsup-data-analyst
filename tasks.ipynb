```python
q6_quantity = 3
q6_unit_price = 2000
q6_discount_rate = 0.15
q6_delivery_fee = 500

q6_subtotal = q6_quantity * q6_unit_price
print(q6_subtotal)
q6_discount_amount = q6_subtotal * q6_discount_rate
print(q6_discount_amount)
q6_customer_payment = q6_subtotal - q6_discount_amount + q6_delivery_fee
print(q6_customer_payment)

assert q6_subtotal == 6000
assert abs(q6_discount_amount - 900) < 1e-9
assert abs(q6_customer_payment - 5600) < 1e-9, "Add delivery fee"
print("Task6 passed")
```
