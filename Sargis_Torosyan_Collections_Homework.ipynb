***Առաջադրանք 1 · List և indexing***

Ստեղծեք q1_products list-ը՝ այս հերթականությամբ՝ "Notebook" , "Pen" , "Bag" , "Pencil" ։

List-ից ստացեք՝

q1_first ՝ առաջին element-ը՝ index-ով։

q1_last ՝ վերջին element-ը՝ negative index-ով։

q1_count ՝ elements-ի քանակը՝ len() -ով, type՝ int ։

q1_has_bag ՝ "Bag" -ի առկայությունը՝ in -ով, type՝ bool ։


```python
q1_products = ["Notebook", "Pen", "Bag", "Pencil"]
q1_first = q1_products[0]
q1_last = q1_products[-1]
q1_count = len(q1_products)
q1_has_bag = "Bag" in q1_products
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q1_products == ["Notebook", "Pen", "Bag", "Pencil"], "Ստուգեք products-ը"
assert q1_first == "Notebook", "Ստուգեք առաջին element-ը"
assert q1_last == "Pencil", "Ստուգեք վերջին element-ը"
assert q1_count == 4, "Ստուգեք elements-ի քանակը"
assert q1_has_bag is True, "Ստուգեք Bag-ի առկայությունը"
print("Առաջադրանք 1․ ստուգումներն անցան")
```

***Առաջադրանք 2 · List-ի թարմացում և ընդլայնում***

Տրված list-ի վրա հերթով կատարեք՝

1. Երկրորդ գինը փոխարինեք 1800 -ով։
2. append() -ով վերջում ավելացրեք 2200 ։
3. extend() -ով ավելացրեք [900, 1100] list-ի երկու elements-ը։
4. q2_count -ում պահեք վերջնական list-ի length-ը։

Բոլոր գները պետք է մնան int ․ վերջնական list-ի մեջ nested list չպետք է լինի։

```python
q2_prices = [1000, 1500, 2000]
q2_prices[1] = 1800
q2_prices.append(2200)
q2_prices.extend([900, 1100])
q2_count = len(q2_prices)
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q2_prices == [1000, 1800, 2000, 2200, 900, 1100], "Ստուգեք գները"
assert q2_count == 6, "Ստուգեք list-ի length-ը"
print("Առաջադրանք 2․ ստուգումներն անցան")
```

***Առաջադրանք 3 · remove և pop***

Գործողությունները կատարեք նշված հերթականությամբ։
1. remove() -ով հեռացրեք առաջին "Pen" -ը։
2. pop() -ով հեռացրեք վերջին element-ը և պահեք q3_removed -ում։
3. q3_pen_count -ում պահեք մնացած "Pen" values-ի քանակը՝ count() -ով։

Վերջնական list-ը պահեք նույն q3_basket անվան տակ։

```python
q3_basket = ["Pen", "Notebook", "Pen", "Bag", "Eraser"]
q3_basket.remove("Pen")
q3_removed = q3_basket.pop()
q3_pen_count = q3_basket.count("Pen")
q3_basket = q3_basket
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q3_basket == ["Notebook", "Pen", "Bag"], "Ստուգեք հեռացումներից հետո մնացած list-ը"  
assert q3_removed == "Eraser", "pop()-ի արդյունքը պահեք q3_removed-ում"
assert q3_pen_count == 1, "Ստուգեք մնացած Pen values-ի քանակը"
print("Առաջադրանք 3․ ստուգումներն անցան")
```

***Առաջադրանք 4 · Թվային list-ի summary***

Տրված non-empty list-ի համար հաշվեք՝

q4_total ՝ գումարը՝ sum() -ով, int ։

q4_average ՝ գումար / քանակ, float ։

q4_minimum և q4_maximum ՝ ամենափոքր ու ամենամեծ values-ը, int ։

q4_sorted ՝ աճման կարգով նոր list՝ sorted() -ով։

q4_amounts սկզբնական list-ի հերթականությունը պահպանեք։ Values-ը
պատվերների գումարներն են՝ AMD-ով։

```python
q4_amounts = [3200, 1800, 4500, 2500]
q4_total = sum(q4_amounts)
q4_average = q4_total / len(q4_amounts)
q4_minimum = min(q4_amounts)
q4_maximum = max(q4_amounts)
q4_sorted = sorted(q4_amounts)
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q4_amounts == [3200, 1800, 4500, 2500], "Սկզբնական list-ը պետք է ճիշտ լինի"
assert q4_total == 12000, "Ստուգեք գումարը"
assert type(q4_average) is float and abs(q4_average - 3000.0) < 1e-9, "Ստուգեք միջինը"
assert (q4_minimum, q4_maximum) == (1800, 4500), "Ստուգեք minimum-ը և maximum-ը"
assert q4_sorted == [1800, 2500, 3200, 4500], "Ստուգեք sorted list-ը"
print("Առաջադրանք 4․ ստուգումներն անցան")
```

***Առաջադրանք 5 · Անկախ copy***

q5_original -ից .copy() -ով ստեղծեք առանձին q5_discounted list։

Միայն copy-ի առաջին գինը նվազեցրեք 200 -ով, ապա copy-ի վերջում
ավելացրեք 4000 ։ Սկզբնական list-ը պետք է մնա անփոփոխ։

q5_original_total -ում և q5_discounted_total -ում պահեք
համապատասխան lists-ի գումարները։

```python
q5_original = [1000, 2000, 3000]
q5_discounted = q5_original.copy()
q5_discounted[0] -= 200
q5_discounted.append(4000)
q5_original_total = sum(q5_original)
q5_discounted_total = sum(q5_discounted)
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q5_original == [1000, 2000, 3000], "Սկզբնական list-ը չպետք է փոխվի"
assert q5_discounted == [800, 2000, 3000, 4000], "Ստուգեք copy-ի փոփոխությունները"
assert q5_discounted is not q5_original, "Ստեղծեք առանձին copy"
assert (q5_original_total, q5_discounted_total) == (6000, 9800), "Ստուգեք sum-ները"
print("Առաջադրանք 5․ ստուգումներն անցան")
```

***Առաջադրանք 6 · Tuple և unpacking***

Տրված tuple-ը unpack արեք երեք variables-ի մեջ՝ q6_product , q6_price ,
q6_currency ։

q6_price -ից հաշվեք q6_revenue ՝ 4 units-ի համար, type՝ int ։

```python
***Առաջադրանք 7 · Dictionary-ի fields-ի փոփոխություն***

q7_product dictionary-ի վրա կատարեք՝

1. "price" key-ի value-ն դարձրեք 1700 ։
2. Ավելացրեք "category" key՝ "Office" value-ով։
3. pop() -ով հեռացրեք "stock" field-ը և նրա value-ն պահեք q7_removed_stock -ում։
4. q7_name -ում պահեք product-ի name-ը՝ dictionary-ից key-ով վերցնելով։

Վերջնական dictionary-ն պետք է ունենա միայն name , price , category fields-ը։
```

***Առաջադրանք 8 · get և membership***

Dictionary-ն մի փոխեք։ Ստացեք՝

q8_city ՝ get() -ով վերցրած city-ի value-ն։

q8_phone ՝ get() -ով phone-ի value-ն, իսկ բացակայության դեպքում՝ "Not provided" ։

q8_has_email ՝ արդյոք "email" key-ը կա, type՝ bool ։

q8_has_ani_value ՝ արդյոք "Ani" value-ն կա dictionary-ի values-ի մեջ, type՝ bool ։

Հիշեք՝ dictionary-ի վրա in -ը ստուգում է keys-ը։

```python
q8_customer = {"name": "Ani", "city": "Yerevan", "email": "ani@example.com"}
q8_city = q8_customer.get("city")
q8_phone = q8_customer.get("phone", "Not provided")
q8_has_email = "email" in q8_customer
q8_has_ani_value = "Ani" in q8_customer.values()
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q8_customer == {"name": "Ani", "city": "Yerevan", "email": "ani@example.com"}, "Ստուգեք սկզբնական dictionary-ը"
assert q8_city == "Yerevan", "Ստուգեք city-ն"
assert q8_phone == "Not provided", "Ստուգեք բացակայող phone-ի default-ը"
assert q8_has_email is True, "Ստուգեք email key-ի առկայությունը"
assert q8_has_ani_value is True, "Ani-ն որոնեք dictionary-ի values-ի մեջ"
print("Առաջադրանք 8․ ստուգումներն անցան")
```

***Առաջադրանք 9 · Nested structures***

Տրված record-ի հետ կատարեք՝

1. q9_first_favorite -ում պահեք առաջին favorite product-ը։
2. q9_email -ում պահեք nested contacts dictionary-ի email-ը։
3. Favorite products list-ի վերջում append() -ով ավելացրեք "Bag" ։
4. Nested contacts-ի city-ն փոխեք "Gyumri" -ի։
5. q9_favorite_count -ում պահեք թարմացված favorite products-ի քանակը։

Մյուս fields-ը պահպանեք։ Օգտագործեք key-երի և index-ի հաջորդական access։

```python
q9_profile = {
    "name": "Aram",
    "favorite_products": ["Notebook", "Pen"],
    "contacts": {"email": "aram@example.com", "city": "Yerevan"}
}

q9_first_favorite = q9_profile["favorite_products"][0]
q9_email = q9_profile["contacts"]["email"]
q9_profile["favorite_products"].append("Bag")
q9_profile["contacts"]["city"] = "Gyumri"
q9_favorite_count = len(q9_profile["favorite_products"])
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q9_first_favorite == "Notebook", "Ստուգեք առաջին favorite producty"
assert q9_email == "aram@example.com", "Ստուգեք nested email-ը"
assert q9_profile == {
    "name": "Aram",
    "favorite_products": ["Notebook", "Pen", "Bag"],
    "contacts": {"email": "aram@example.com", "city": "Gyumri"}
}, "Ստուգեք profile-ի փոփոխությունները"
assert q9_favorite_count == 3, "Ստուգեք թարմացված favorites-ի քանակը"
print("Առաջադրանք 9․ ստուգումներն անցան")
```

***Առաջադրանք 10 · Փոքր dataset և ութ analytical հարց***

Յուրաքանչյուր dictionary-ն մեկ ապրանք պարունակող պատվեր է։ Երկու records-ին ավելացրեք "revenue" field՝ quantity × unit price հաշվարկով։ Մնացած fields-ը պահպանեք։

Ստեղծեք q10_revenues list՝ առաջին և երկրորդ պատվերների հաշվված revenues-ով՝ նույն հերթականությամբ։ Այնուհետև ստեղծեք q10_summary dictionary՝ ճիշտ այս ութ keys-ով։

Key             | Հարց կամ պահանջ                     | Type
____________________________________________________________
order_count     | Քանի՞ պատվեր կա dataset-ում         | int

first_product   | Ո՞րն է առաջին պատվերի product-ը     | str

second_quantity | Որքա՞ն է երկրորդ պատվերի quantity-ն | int

first_revenue   | Որքա՞ն է առաջին պատվերի revenue-ն   | int

second_revenue  | Որքա՞ն է երկրորդ պատվերի revenue-ն  | int

total_revenue   | Որքա՞ն է ընդհանուր revenue-ն        | int

average_revenue | Որքա՞ն է average order revenue-ն    | float

max_revenue     | Որքա՞ն է ամենամեծ պատվերի revenue-ն | int




Օգտագործեք dataset-ից ստացված values-ը և len() , sum() , max() ։

Average-ը հաշվեք total / order count։

Այս երկու records-ի համար calculations-ը գրեք առանձին՝ indexes-ով։

Լուծման վերջում print() -երով ցուցադրեք summary-ն և comment-ով գրեք մեկ observation՝ թվով ու AMD unit-ով։

Օրինակում discount, delivery կամ returns չկան։

```python
q10_orders = [
    {"order_id": "B101", "product": "Notebook", "quantity": 4, "unit_price": 1250},
    {"order_id": "B102", "product": "Pen", "quantity": 3, "unit_price": 800}
]

q10_orders[0]["revenue"] = q10_orders[0]["quantity"] * q10_orders[0]["unit_price"]
q10_orders[1]["revenue"] = q10_orders[1]["quantity"] * q10_orders[1]["unit_price"]
q10_revenues = [q10_orders[0]["revenue"], q10_orders[1]["revenue"]]
q10_total_revenue = sum(q10_revenues)
q10_revenues_count = len(q10_revenues)

q10_summary = {
    "order_count": len(q10_orders),
    "first_product": q10_orders[0]["product"],
    "second_quantity": q10_orders[1]["quantity"],
    "first_revenue": q10_revenues[0],
    "second_revenue": q10_revenues[1],
    "total_revenue": q10_total_revenue,
    "average_revenue": q10_total_revenue / q10_revenues_count,
    "max_revenue": max(q10_revenues)
}

print(q10_summary)

#Observatioin: there are 2 orders with total revenue 7400. The max revenue is 5000
```

```python
# Գործարկեք ձեր լուծումից հետո։
assert q10_orders == [
    {"order_id": "B101", "product": "Notebook", "quantity": 4, "unit_price": 1250, "revenue": 5000},
    {"order_id": "B102", "product": "Pen", "quantity": 3, "unit_price": 800, "revenue": 2400}
], "Ստուգեք պատվերների fields-ը և հաշվված revenues-ը"
assert q10_revenues == [5000, 2400], "Ստուգեք revenues list-ը"
assert q10_summary == {
    "order_count": 2,
    "first_product": "Notebook",
    "second_quantity": 3,
    "first_revenue": 5000,
    "second_revenue": 2400,
    "total_revenue": 7400,
    "average_revenue": 3700.0,
    "max_revenue": 5000
}, "Ստուգեք summary-ի ութ fields-ը և դրանց values-ը"
assert type(q10_summary["average_revenue"]) is float, "Average revenue-ն սխալ է"
print("Առաջադրանք 10․ ստուգումներն անցան")
```
