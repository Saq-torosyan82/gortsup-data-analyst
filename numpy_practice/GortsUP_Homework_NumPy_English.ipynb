## 10. Calculate the Summary for the First Group

Select observations where `cultivar_id == 1`. Calculate the group size and the means of the three columns. Solve the same task using a small function as well. For a code that is not found, return `(0, None)`.


### Code


```python
# Write your solution here
def group_summary(X, cultivar_id, code):
    selected = X[cultivar_id == code]
    if selected.shape[0] == 0:
        return (0, None)
    return (selected.shape[0], selected.mean(axis=0))

cultivar_id = y
cultivar_id_1 = X[cultivar_id == 1]
count_1, means_1 = group_summary(X, cultivar_id, 1)
missing_count, missing_means = group_summary(X, cultivar_id, 99)
print(count_1)
print(means_1)
print(missing_count, missing_means)
```

## 14. Handle an Empty Selection

Write a `safe_alcohol_mean(X, threshold)` function. If there are no selected observations, return `(0, None)`. Test it with thresholds 13 and 100.


### Code


```python
# Write your solution here
def safe_alcohol_mean(X, threshold):
    selected = X[X[:, 0] > threshold, 0]
    if len(selected) == 0:
        return (0, None)
    return (len(selected), float(selected.mean()))

result_13 = safe_alcohol_mean(X, 13)
result_100 = safe_alcohol_mean(X, 100)
print(result_13)
print(result_100)
```

### Checks


```python
# Write your assertions / checks here
assert result_13[0] == 92
assert np.isclose(result_13[1], X[X[:, 0] > 13, 0].mean())
assert result_100 == (0, None)
print("passed")
```

### Short English Comment

- Threshold 13 selects 92 alcohol values, with mean 13.67.
- Threshold 100 selects nothing, so the function returns `(0, None)`.
- The empty case avoids calling the mean on an empty array.
