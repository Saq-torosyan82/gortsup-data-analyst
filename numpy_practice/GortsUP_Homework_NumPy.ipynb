## Աշխատանքի մեկնարկը


```python
import numpy as np
from pathlib import Path

path = Path("data") / "wine_selected.csv"
raw = np.loadtxt(path, delimiter=",", skiprows=1)
y = raw[:, 0].astype(np.int64)
X = raw[:, 1:].copy()
assert X.shape == (178, 3)
assert np.isfinite(X).all()
print("passed")
```

## 1. Բեռնեք և առանձնացրեք տվյալները

Բեռնեք `data/wine_selected.csv` ֆայլը։ Առանձնացրեք `cultivar_id`-ը որպես ամբողջ թվերի `y` զանգված, իսկ մյուս երեք սյունակները՝ որպես իրական թվերի `X` զանգված։ Տպեք `X`-ի և `y`-ի ձևերը, `X`-ի առանցքների և տարրերի քանակը։ Բացատրեք՝ ինչ է ներկայացնում մեկ տողը։

**Ինքնաստուգում։** `X.shape == (178, 3)`, `y.shape == (178,)`։


### Կոդ


```python
# Գրեք այստեղ լուծումը
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert X.shape == (178, 3)
assert y.shape == (178,)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

Ֆայլը ունի 178 տող և 4 սյուն։

## 2. Ստուգեք տիպը և տասնորդական մասը

Տպեք `X.dtype`-ը։ Առաջին հինգ դիտարկումների ալկոհոլի ցուցանիշները փոխարկեք ամբողջ թվերի՝ առանձին զանգվածում։ Համեմատեք սկզբնականների հետ։ Գրեք՝ ինչ տեղեկություն է կորում, և ինչու վերլուծության համար պահում եք սկզբնական իրական թվերը։ `X`-ը մի փոխեք։


### Կոդ


```python
# Գրեք այստեղ լուծումը
print(X.dtype)
alcohol_column = X[:5, 0]
alcohol_column_int = alcohol_column.astype(np.int64)
print(alcohol_column)
print(alcohol_column_int)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert alcohol_column.shape == (5,)
assert alcohol_column_int.dtype == np.int64
assert alcohol_column_int.shape == (5,)
assert np.array_equal(alcohol_column_int, np.trunc(alcohol_column).astype(np.int64))
assert not np.allclose(alcohol_column, alcohol_column_int)
assert X.dtype == np.float64
assert np.allclose(X[:5, 0], alcohol_column)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- `X.dtype`-ը `float64` է։

- Ալկոհոլի առաջին հինգ արժեքներն են՝ `14.23`, `13.2`, `13.16`, `14.37` և `13.24`։

- Դրանք `int64` տիպի վերածելիս կոտորակային մասը հեռացվում է (կտրվում), ուստի արժեքները դառնում են `14`, `13`, `13`, `14` և `13`։

- Կորսված այդ կոտորակային մասը չափման մի մասն է կազմում, և ամբողջ թվերի օգտագործումը կնվազեցնի հետագա միջինների և տարբերությունների ճշգրտությունը։

- `X`-ում առկա սկզբնական `float64` արժեքները մնում են անփոփոխ։

## 3. Ընտրեք մեկ դիտարկում և մեկ հատկանիշ

Ստացեք առաջին դիտարկման բոլոր երեք չափումները, վերջին դիտարկման ալկոհոլի ցուցանիշը և մագնեզիումի ամբողջ սյունակը։ Գրեք յուրաքանչյուր արդյունքի ձևը։


### Կոդ


```python
# Գրեք այստեղ լուծումը
first_row = X[0]
last_row = X[-1]
last_row_alcohol = last_row[0]
magnesium_column = X[:, 2]
print(first_row, first_row.shape)
print(last_row_alcohol, last_row_alcohol.shape)
print(magnesium_column.shape)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert first_row.shape == (3,)
assert last_row.shape == (3,)
assert last_row_alcohol.dtype == np.float64
assert magnesium_column.dtype == np.float64
assert magnesium_column.shape == (178,)
assert np.allclose(last_row_alcohol, last_row[0])
assert np.allclose(magnesium_column, X[:, 2])
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Առաջին դիտարկումը պարունակում է բոլոր երեք չափումները, ուստի `first_row.shape`-ը `(3,)` է։

- Վերջին դիտարկման ալկոհոլի ցուցանիշը մեկ թիվ է, ուստի `last_row_alcohol.shape`-ը `()` է։

- Մագնեզիումի սյունակը յուրաքանչյուր 178 դիտարկման համար ունի մեկական արժեք, ուստի `magnesium_column.shape`-ը `(178,)` է։


## 4. Ստեղծեք ուսումնասիրվող հատված

Ընտրեք առաջին տասներկու դիտարկումների ալկոհոլի և խնձորաթթվի սյունակները՝ անկախ պատճենով։ Ստուգեք ձևը և ապացուցեք, որ պատճենի փոփոխությունը չի փոխում `X`-ը։


### Կոդ


```python
# Գրեք այստեղ լուծումը
study_subset = X[:12, :2].copy()
print(study_subset.shape)
x_before = X.copy()
study_subset[0, 0] = -1
print(np.array_equal(X, x_before))
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert study_subset.shape == (12, 2)
assert not np.shares_memory(study_subset, X)
assert study_subset[0, 0] == -1
assert np.array_equal(X, x_before)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- `X[:12, :2]`-ը ընտրում է ալկոհոլի և խնձորաթթվի արժեքները առաջին 12 դիտարկումների համար։

- Առանց `.copy()`-ի՝ այդ ընտրված հատվածը հանդիսանում է տվյալների «տեսք» (view), ուստի դրանում կատարված փոփոխությունները կազդեն նաև `X`-ի վրա։

- `.copy()`-ն ստեղծում է անկախ զանգված՝ `(12, 2)` չափսով։

- `study_subset[0, 0]`-ին `-1` արժեքը վերագրելու դեպքում `X`-ը մնում է անփոփոխ։

## 5. Համեմատեք երկու հաշվարկման մոտեցում

Հաշվեք ալկոհոլի ցուցանիշի ընդհանուր գումարը բացահայտ `for` ցիկլով և NumPy-ի `sum()`-ով։ `np.isclose()`-ով ստուգեք համընկնումը։


### Կոդ


```python
# Գրեք այստեղ լուծումը
total_alcohol = 0
for row in X:
    total_alcohol += row[0]
print(total_alcohol)
total_alcohol_np = np.sum(X[:, 0])
print(total_alcohol_np)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert np.isclose(total_alcohol, total_alcohol_np)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Երկու հաշվարկներն էլ գումարում են ալկոհոլի սյունակը՝ `X[:, 0]`-ը։

- Ցիկլի միջոցով ստացված ընդհանուր գումարը `2314.1099999999988` է, իսկ `np.sum`-ը տալիս է `2314.11` արժեքը։

- Այս չնչին տարբերությունը պայմանավորված է լողացող կետով թվերի գումարման հերթականությամբ, ուստի `np.isclose`-ը ճիշտ է արձանագրում, որ երկու արդյունքները համընկնում են։


## 6. Ամփոփեք երեք հատկանիշները

Յուրաքանչյուր սյունակի համար հաշվեք նվազագույնը, առավելագույնը և միջինը։ Կազմեք երեք տողանոց ընթեռնելի ամփոփում։


### Կոդ


```python
# Գրեք այստեղ լուծումը
min_alcohol = np.min(X[:, 0])
max_alcohol = np.max(X[:, 0])
mean_alcohol = np.mean(X[:, 0])
min_malic = np.min(X[:, 1])
max_malic = np.max(X[:, 1])
mean_malic = np.mean(X[:, 1])
min_magnesium = np.min(X[:, 2])
max_magnesium = np.max(X[:, 2])
mean_magnesium = np.mean(X[:, 2])
print(f"alcohol: min={min_alcohol:.2f}, max={max_alcohol:.2f}, mean={mean_alcohol:.2f}")
print(f"malic acid: min={min_malic:.2f}, max={max_malic:.2f}, mean={mean_malic:.2f}")
print(f"magnesium: min={min_magnesium:.2f}, max={max_magnesium:.2f}, mean={mean_magnesium:.2f}")
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert np.allclose(min_alcohol, np.min(X[:, 0]))
assert np.allclose(max_alcohol, np.max(X[:, 0]))
assert np.allclose(mean_alcohol, np.mean(X[:, 0]))
assert np.allclose(min_malic, np.min(X[:, 1]))
assert np.allclose(max_malic, np.max(X[:, 1]))
assert np.allclose(mean_malic, np.mean(X[:, 1]))
assert np.allclose(min_magnesium, np.min(X[:, 2]))
assert np.allclose(max_magnesium, np.max(X[:, 2]))
assert np.allclose(mean_magnesium, np.mean(X[:, 2]))
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Ալկոհոլի պարունակությունը տատանվում է 11,03-ից մինչև 14,83՝ 13,00 միջին արժեքով։
- Խնձորաթթվի պարունակությունը տատանվում է 0,74-ից մինչև 5,80՝ 2,34 միջին արժեքով։
- Մագնեզիումի պարունակությունը տատանվում է 1,36-ից մինչև 3,23՝ 2,37 միջին արժեքով։
- Յուրաքանչյուր տող ներկայացնում է մեկ հատկանիշ՝ իր նվազագույն, առավելագույն և միջին արժեքներով։


## 7. Գտեք փոփոխման տիրույթը

Յուրաքանչյուր հատկանիշի համար առավելագույնից հանեք նվազագույնը։ Ստացեք երեք արժեք և բացատրեք՝ ինչու տարբեր սանդղակներով տիրույթների մեծությունները ուղղակի համեմատելը բավարար չէ։


### Կոդ


```python
# Գրեք այստեղ լուծումը
range_alcohol = max_alcohol - min_alcohol
range_malic = np.max(X[:, 1]) - np.min(X[:, 1])
range_magnesium = np.max(X[:, 2]) - np.min(X[:, 2])
print(f"alcohol: {range_alcohol:.2f}")
print(f"malic acid: {range_malic:.2f}")
print(f"magnesium: {range_magnesium:.2f}")
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert range_alcohol.dtype == np.float64
assert range_malic.dtype == np.float64
assert range_magnesium.dtype == np.float64
assert np.allclose(range_alcohol, max_alcohol - min_alcohol)
assert np.allclose(range_malic, np.max(X[:, 1]) - np.min(X[:, 1]))
assert np.allclose(range_magnesium, np.max(X[:, 2]) - np.min(X[:, 2]))
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Ալկոհոլի միջակայքը 3,80 է՝ 11,03-ից մինչև 14,83։
- Խնձորաթթվի միջակայքը 5,06 է՝ 0,74-ից մինչև 5,80։
- Մագնեզիումի միջակայքը 1,87 է՝ 1,36-ից մինչև 3,23։
- Ավելի մեծ միջակայքը չի նշանակում ավելի մեծ փոփոխականություն, քանի որ հատկանիշները չափվում են տարբեր սանդղակներով։ Խնձորաթթվի միջակայքն ավելի մեծ է, քան ալկոհոլինը, սակայն ալկոհոլի արժեքները կենտրոնացած են 13-ի, իսկ խնձորաթթվի արժեքները՝ 2-ի մոտակայքում։


## 8. Կազմեք մեկ պայմանով ընտրություն

Ընտրեք այն դիտարկումները, որոնց ալկոհոլի ցուցանիշը խիստ մեծ է 13-ից։ Հաշվեք քանակը և տոկոսը։ Վերադարձրեք նաև ընտրված `X` և համապատասխան `y`։


### Կոդ


```python
# Գրեք այստեղ լուծումը
alcohol_gt_13 = X[X[:, 0] > 13]
y_alcohol_gt_13 = y[X[:, 0] > 13]
count_alcohol_gt_13 = len(alcohol_gt_13)
percent_alcohol_gt_13 = 100 * count_alcohol_gt_13 / len(X)
print(count_alcohol_gt_13)
print(f"{percent_alcohol_gt_13:.2f}%")
print(alcohol_gt_13.shape, y_alcohol_gt_13.shape)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert alcohol_gt_13.shape == (92, 3)
assert y_alcohol_gt_13.shape == (92,)
assert np.all(alcohol_gt_13[:, 0] > 13)
assert count_alcohol_gt_13 == 92
assert np.isclose(percent_alcohol_gt_13, 100 * 92 / 178)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- 178 դիտարկումներից 92-ում ալկոհոլի պարունակությունը խստորեն գերազանցում է 13-ը։
- Դա կազմում է տողերի 51,69%-ը։
- `alcohol_gt_13`-ն ունի `(92, 3)` չափականություն, իսկ համապատասխան `y_alcohol_gt_13` պիտակները՝ `(92,)` չափականություն։


## 9. Միացրեք պայմանները

Ընտրեք դիտարկումները, որոնց ալկոհոլը > 13 և մագնեզիումը >= 100։ Օգտագործեք փակագծեր և `&`։ Հաշվեք քանակը և ընտրված խմբի խնձորաթթվի միջինը։


### Կոդ


```python
# Գրեք այստեղ լուծումը
selected_mask = (X[:, 0] > 13) & (X[:, 2] >= 100)
alcohol_gt_13_and_magnesium_ge_100 = X[selected_mask]
selected_count = int(selected_mask.sum())
if selected_count == 0:
    selected_malic_mean = None
else:
    selected_malic_mean = alcohol_gt_13_and_magnesium_ge_100[:, 1].mean()
print(selected_count)
print(selected_malic_mean)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert alcohol_gt_13_and_magnesium_ge_100.shape == (0, 3)
assert selected_count == 0
assert selected_malic_mean is None
assert np.array_equal(selected_mask, (X[:, 0] > 13) & (X[:, 2] >= 100))
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Յուրաքանչյուր համեմատություն փակագծերի մեջ է, իսկ `&` նշանը միավորում է երկու դիմակները (masks)։
- Ոչ մի տողում ալկոհոլի պարունակությունը 13-ից բարձր չէ, իսկ մագնեզիումինը՝ 100-ից ոչ պակաս։
- Քանակը 0 է, ուստի խնձորաթթվի (malic acid) միջին արժեքը սահմանված չէ։


## 10. Հաշվարկեք առաջին խմբի ամփոփումը

Ընտրեք `cultivar_id == 1` դիտարկումները։ Հաշվեք խմբի քանակը և երեք սյունակների միջինները։ Նույնը լուծեք նաև փոքր ֆունկցիայով։ Չգտնված կոդի համար վերադարձեք `(0, None)`։


### Կոդ


```python
# Գրեք այստեղ լուծումը
def group_summary(X, cultivar_id, code):
    selected = X[cultivar_id == code]
    if selected.shape[0] == 0:
        return (0, None)
    return (selected.shape[0], selected.mean(axis=0))

cultivar_id = y
cultivar_id_1 = X[cultivar_id == 1]
count_1, means_1 = group_summary(X, cultivar_id, 1)
missing_count, missing_means = group_summary(X, cultivar_id, 99)
print(count_1)
print(means_1)
print(missing_count, missing_means)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert cultivar_id_1.shape == (59, 3)
assert count_1 == 59
assert np.allclose(means_1, cultivar_id_1.mean(axis=0))
assert (missing_count, missing_means) == (0, None)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- 1-ին սորտն ունի 59 դիտարկում։
- Դրա միջին արժեքներն են՝ ալկոհոլ՝ 13,74, խնձորաթթու՝ 2,01 և մագնեզիում՝ 2,46։
- Երբ սորտի կոդը բացակայում է, `group_summary`-ն վերադարձնում է `(0, None)`։

## 11. Համեմատեք բոլոր խմբերը

1, 2, 3 խմբերի համար կազմեք քանակների զանգված և սյունակների միջինների `(3, 3)` աղյուսակ։ `np.argmax()`-ով գտեք ալկոհոլի ամենամեծ միջինն ունեցող խմբի դիրքը և այն վերածեք խմբի կոդի։


### Կոդ


```python
# Գրեք այստեղ լուծումը
group_codes = np.array([1, 2, 3])
group_counts = np.array([(y == 1).sum(), (y == 2).sum(), (y == 3).sum()])
group_means = np.array([
    X[y == 1].mean(axis=0),
    X[y == 2].mean(axis=0),
    X[y == 3].mean(axis=0),
])
highest_alcohol_position = int(np.argmax(group_means[:, 0]))
highest_alcohol_group = int(group_codes[highest_alcohol_position])
print(group_counts)
print(group_means)
print(highest_alcohol_position, highest_alcohol_group)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert group_counts.shape == (3,)
assert np.array_equal(group_counts, np.array([59, 71, 48]))
assert group_means.shape == (3, 3)
assert highest_alcohol_position == 0
assert highest_alcohol_group == 1
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Խմբերի քանակներն են՝ 59, 71 և 48։
- `group_means`-ը սյունակների միջին արժեքներից բաղկացած `(3, 3)` չափսի աղյուսակ է։
- `np.argmax`-ը վերադարձնում է 0 դիրքը, որը համապատասխանում է 1-ին սորտի կոդին և ալկոհոլի ամենաբարձր միջին ցուցանիշին։


## 12. Հաշվեք կշռված ընդհանուր միջինը

Օգտագործեք նախորդ առաջադրանքի խմբային քանակներն ու ալկոհոլի միջինները։ Հաշվեք պարզ միջինը և քանակներով կշռված միջինը։


### Կոդ


```python
# Գրեք այստեղ լուծումը
simple_alcohol_mean = group_means[:, 0].mean()
weighted_alcohol_mean = np.sum(group_means[:, 0] * group_counts) / group_counts.sum()
print(simple_alcohol_mean)
print(weighted_alcohol_mean)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert np.isclose(simple_alcohol_mean, group_means[:, 0].mean())
assert np.isclose(weighted_alcohol_mean, X[:, 0].mean())
assert not np.isclose(simple_alcohol_mean, weighted_alcohol_mean)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- Երեք խմբերի՝ ալկոհոլի ցուցանիշների միջին թվաբանականը 13.06 է։
- Ըստ քանակի կշռված միջինը 13.00 է։
- Կշռված միջինը համընկնում է ալկոհոլի ընդհանուր միջին ցուցանիշի հետ, քանի որ ավելի մեծ խմբերն ավելի մեծ կշիռ ունեն։

## 13. Հաշվեք միջինից շեղումները

Յուրաքանչյուր արժեքից հանեք իր սյունակի միջինը՝ առանց տողերով ցիկլի։ Արդյունքը պահեք `centered` փոփոխականում։ Ստուգեք `np.allclose(centered.mean(axis=0), 0, atol=1e-12)`։


### Կոդ


```python
# Գրեք այստեղ լուծումը
centered = X - X.mean(axis=0)
print(centered.shape)
print(centered.mean(axis=0))
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert centered.shape == X.shape
assert np.allclose(centered.mean(axis=0), 0, atol=1e-12)
assert np.allclose(centered, X - X.mean(axis=0))
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- `X.mean(axis=0)`-ի հանումը կենտրոնացնում է յուրաքանչյուր սյունակ՝ առանց տողերով ցիկլի կիրառման։
- `centered`-ը պահպանում է սկզբնական չափսը՝ `(178, 3)`։
- Յուրաքանչյուր սյունակի միջին արժեքը 0 է՝ `1e-12` թույլատրելի շեղման սահմաններում։

## 14. Մշակեք դատարկ ընտրությունը

Գրեք `safe_alcohol_mean(X, threshold)` ֆունկցիան։ Եթե ընտրվածներ չկան, վերադարձեք `(0, None)`։ Փորձարկեք 13 և 100 շեմերով։


### Կոդ


```python
# Գրեք այստեղ լուծումը
def safe_alcohol_mean(X, threshold):
    selected = X[X[:, 0] > threshold, 0]
    if len(selected) == 0:
        return (0, None)
    return (len(selected), float(selected.mean()))

result_13 = safe_alcohol_mean(X, 13)
result_100 = safe_alcohol_mean(X, 100)
print(result_13)
print(result_100)
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert result_13[0] == 92
assert np.isclose(result_13[1], X[X[:, 0] > 13, 0].mean())
assert result_100 == (0, None)
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- 13 շեմը ընտրում է ալկոհոլի 92 արժեք, որոնց միջինը 13.67 է։
- 100 շեմը ոչինչ չի ընտրում, ուստի ֆունկցիան վերադարձնում է `(0, None)`։
- Դատարակ դեպքը թույլ է տալիս խուսափել դատարկ զանգվածի համար միջին արժեքը հաշվարկելուց։

## 15. Կազմեք վերարտադրելի փոքր հաշվետվություն

Բառարանում պահեք ընդհանուր դիտարկումների քանակը, ալկոհոլի միջինը, 13 շեմը գերազանցողների քանակն ու տոկոսը և խմբերի քանակները։ Ավելացրեք առնվազն երեք `assert`, երեք եզրակացություն, մեկ սահմանափակում և մեկ հաջորդ հարց։


### Կոդ


```python
# Գրեք այստեղ լուծումը
n_observations = len(X)
alcohol_mean = float(X[:, 0].mean())
alcohol_gt_13_count = int((X[:, 0] > 13).sum())
alcohol_gt_13_percent = float(100 * alcohol_gt_13_count / n_observations)
report = {
    "n_observations": n_observations,
    "alcohol_mean": alcohol_mean,
    "alcohol_gt_13_count": alcohol_gt_13_count,
    "alcohol_gt_13_percent": alcohol_gt_13_percent,
    "group_counts": {
        1: int((y == 1).sum()),
        2: int((y == 2).sum()),
        3: int((y == 3).sum()),
    },
    "conclusions": [
        "92 observations, 51.69%, have alcohol above 13.",
        "Group 2 is the largest group, with 71 observations, but it has the lowest mean alcohol.",
        "Group 1 has the highest mean alcohol.",
    ],
    "limitation": "The magnesium column only ranges from 1.36 to 3.23, so magnesium >= 100 selects no rows.",
    "follow_up_question": "Would the group ranking stay the same if the comparison used malic acid instead of alcohol?",
}
print(report["n_observations"], report["alcohol_mean"])
print(report["alcohol_gt_13_count"], report["alcohol_gt_13_percent"])
print(report["group_counts"])
```

### Ստուգումներ


```python
# Գրեք այստեղ assert-ները / ստուգումները
assert report["n_observations"] == 178
assert np.isclose(report["alcohol_mean"], X[:, 0].mean())
assert report["alcohol_gt_13_count"] == 92
assert np.isclose(report["alcohol_gt_13_percent"], 100 * 92 / 178)
assert report["group_counts"] == {1: 59, 2: 71, 3: 48}
assert len(report["conclusions"]) >= 3
print("passed")
```

### Կարճ հայերեն մեկնաբանություն

- 92 դիտարկման դեպքում (51,69%) ալկոհոլի պարունակությունը 13-ից բարձր է։
- 2-րդ խումբն ամենամեծն է՝ 71 դիտարկմամբ, սակայն այնտեղ ալկոհոլի միջին ցուցանիշն ամենացածրն է։
- 1-ին խմբում ալկոհոլի միջին ցուցանիշն ամենաբարձրն է։
- Սահմանափակում․ մագնեզիումի սյունակում արժեքները տատանվում են միայն 1,36-ից 3,23-ի սահմաններում, ուստի «մագնեզիում >= 100» պայմանը կիրառելիս ոչ մի տող չի ընտրվում։
