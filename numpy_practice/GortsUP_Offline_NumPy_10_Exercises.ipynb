## Ինչպես աշխատել

1. Ձեր լուծման բջիջում `None`-ի փոխարեն գրեք անհրաժեշտ արտահայտությունը։
2. Գործարկեք Ձեր լուծման բջիջը, ապա ստուգումների բջիջը։
3. Եթե ստուգումն անցնում է, կտեսնեք հաջողության հաղորդագրություն։ Եթե ստանում եք `AssertionError`, վերադարձեք Ձեր լուծմանը։
4. Առաջարկվող լուծումը դիտեք Ձեր փորձից և քննարկումից հետո։ Այն գործարկելուց հետո կրկին գործարկեք նույն ստուգումները։

Բոլոր վարժությունները միանգամից մի գործարկեք, քանի դեռ լուծումները չեք լրացրել։ Այս դասում նոր ֆունկցիաներ սահմանելու կարիք չկա։ Նախ գործարկեք տվյալների պատրաստման բջիջը։

## Կրկնենք․ ցուցակ և NumPy զանգված

Ցուցակի և NumPy զանգվածի գրառումները նման են, սակայն նույն գործողությունը կարող է տարբեր իմաստ ունենալ։

Ցուցակը թվով բազմապատկելիս ցուցակի պարունակությունը կրկնվում է։ Թվային NumPy զանգվածը թվով բազմապատկելիս **յուրաքանչյուր թիվ է բազմապատկվում**։

Ստեղծելու համար ցուցակը փոխանցում ենք `np.array()`-ին։ Հաջորդ փոքր օրինակը ուսուցողական է։ Նախ կանխատեսեք երկու արդյունքները, հետո գործարկեք։

```python
numbers_list = [2, 4, 6]
numbers_array = np.array(numbers_list)

print("Ցուցակի դեպքում՝", numbers_list * 2)
print("Զանգվածի դեպքում՝", numbers_array * 2)
```

## Կրկնենք․ ի՞նչ է ցույց տալիս shape-ը

`X.shape`-ը ցույց է տալիս տողերի և սյունակների քանակը։ Մեր դեպքում արդյունքը `(150, 4)` է։

- `X.shape[0]`՝ տողերի, այսինքն՝ ծաղիկների քանակը։
- `X.shape[1]`՝ սյունակների, այսինքն՝ յուրաքանչյուր ծաղկի չափումների քանակը։

`X.dtype`-ը ցույց է տալիս թվերի տիպը։ Չափումներն ունեն տասնորդական մաս, ուստի դրանք պահում ենք որպես իրական թվեր։ Այդ թվերը ամբողջների փոխարկելու կարիք չկա։

```python
print("Տողեր և սյունակներ՝", X.shape)
print("Ծաղիկների քանակը՝", X.shape[0])
print("Չափումների քանակը՝", X.shape[1])
print("Տվյալների տիպը՝", X.dtype)
```

## Կրկնենք․ մեկ տող մեկ սյունակ կամ փոքր հատված

Ինդեքսները սկսվում են **0-ից**։ Ուստի առաջին տողը `X[0]`-ն է։

| Գրառումը | Ինչ է ընտրում |
|---|---|
| `X[0]` | Առաջին ծաղկի բոլոր չափումները |
| `X[0, 2]` | Առաջին ծաղկի պսակաթերթի երկարությունը |
| `X[:, 2]` | Բոլոր ծաղիկների պսակաթերթերի երկարությունները |
| `X[:3, :2]` | Առաջին երեք տողի առաջին երկու սյունակները |

`:` նշանը՝ առանց թվերի, նշանակում է տվյալ ուղղության բոլոր դիրքերը։ `:3`-ը ընտրում է 0, 1, 2 դիրքերը․ աջ սահմանը չի ներառվում։

Սկզբում միշտ հարցրեք՝ ուզում եմ մեկ ծաղկի՞ տվյալները, թե՞ բոլոր ծաղիկների մեկ չափումը։

```python
print("Առաջին ծաղիկը՝", X[0])
print("Առաջին ծաղկի պսակաթերթի երկարությունը՝", X[0, 2])
print("Առաջին երեք տողի առաջին երկու սյունակները՝")
print(X[:3, :2])
```

## Հաշվարկ ամբողջ սյունակի համար

Մեկ սանտիմետրը տասը միլիմետր է։ Ուստի սանտիմետրով տրված թվերը միլիմետրի վերածելու համար բազմապատկում ենք 10-ով։ NumPy-ն այդ գործողությունը կարող է կատարել ամբողջ սյունակի համար։

Ստորև պսակաթերթերի լայնություններն ենք ընտրում՝ 3 համարի սյունակը։ `* 10`-ի արդյունքը պահում ենք նոր փոփոխականում։ Սկզբնական `X`-ը չի փոխվում։

```python
widths_cm = X[:, 3]
widths_mm = widths_cm * 10

print("Առաջին հինգ լայնությունները սանտիմետրերով՝", widths_cm[:5])
print("Նույն լայնությունները միլիմետրերով՝", widths_mm[:5])
```

## Միջին նվազագույն և առավելագույն արժեքները

Երբ ունենք մեկ չափման ամբողջ սյունակը, կարող ենք այն ամփոփել։

- `.mean()`՝ միջին արժեքը։
- `.min()`՝ ամենափոքր արժեքը։
- `.max()`՝ ամենամեծ արժեքը։

Ստորև աշխատում ենք **բաժակաթերթի երկարության** հետ։ Արդյունքների միավորը մնում է սանտիմետրը։ Նախ փորձեք բացատրել՝ ինչու միջինը պետք է լինի նվազագույնի և առավելագույնի միջև։

```python
sepal_lengths = X[:, 0]
print("Բաժակաթերթի երկարության միջինը՝", sepal_lengths.mean())
print("Ամենափոքր երկարությունը՝", sepal_lengths.min())
print("Ամենամեծ երկարությունը՝", sepal_lengths.max())
```

## Նոր քայլ․ ընտրենք մեկ պայմանով

Ենթադրենք՝ ուզում ենք գտնել 6 սմ-ից մեծ բաժակաթերթի երկարություն ունեցող ծաղիկները։ Նախ գրում ենք պայմանը՝ `X[:, 0] > 6`։

Այդ արտահայտության արդյունքը `True` և `False` արժեքների զանգված է։ Այն ցույց է տալիս՝ յուրաքանչյուր ծաղիկը բավարարո՞ւմ է պայմանին։ Այսպիսի զանգվածը կոչվում է **դիմակ**։

`X[mask]`-ը պահում է միայն այն տողերը, որոնց համար պայմանը ճիշտ է։ Ընտրվածների քանակը կարող ենք գտնել `selected.shape[0]`-ով։ Այս դասում օգտագործում ենք միայն մեկ պայման։

```python
mask = X[:, 0] > 6.0
selected = X[mask]

print("Առաջին հինգ տողի համար պայմանը ճիշտ է՝", mask[:5])
print("Ընտրված ծաղիկների քանակը՝", selected.shape[0])
print("Ընտրված առաջին երեք տողերը՝")
print(selected[:3])
```

## Նոր քայլ․ յուրաքանչյուր սյունակի միջինը

Եթե ուզում ենք չորս տարբեր հատկանիշների չորս առանձին միջինները, գրում ենք `X.mean(axis=0)`։

**Այս դասի համար հիշենք պարզ կանոնը․ `axis=0`-ով ստանում ենք մեկ արդյունք յուրաքանչյուր սյունակի համար։**

Սկզբում նայենք փոքր ուսուցողական աղյուսակին։ Առաջին սյունակի միջինը 15 է, երկրորդինը՝ 150։ Հետո նույն գործողությունը կանենք իրական տվյալներով։ Տարբեր հատկանիշների բոլոր թվերը մեկ միջինի մեջ չենք խառնում։

```python
small_example = np.array([[10, 100], [20, 200]])
print("Փոքր աղյուսակը՝")
print(small_example)
print("Յուրաքանչյուր սյունակի միջինը՝", small_example.mean(axis=0))
```

## Ինչպես են աշխատում ստուգումները

`assert`-ով գրում ենք մի պայման, որը ճիշտ լուծման դեպքում պետք է կատարվի։ Եթե պայմանը սխալ է, կտեսնենք `AssertionError`։ Դա օգնում է գտնել սխալը։

Ամբողջ թվերի համար կարող ենք օգտագործել `==`։ Տասնորդական թվերը համեմատելու համար հարմար է `np.isclose()`-ը, իսկ թվերի ամբողջ զանգվածների համար՝ `np.allclose()`-ը։ Այդպես հաշվարկի շատ փոքր կլորացման տարբերությունը սխալ չենք համարի։

Հաջորդ օրինակը միայն ստուգումների ցուցադրություն է։ Վարժություններում ստուգումներն արդեն գրված են։

```python
example_count = 3
example_mean = np.array([1.0, 2.0, 3.0]).mean()

assert example_count == 3
assert np.isclose(example_mean, 2.0)
assert np.allclose(np.array([1.0, 2.0]) * 10, [10.0, 20.0])
print("Օրինակի ստուգումներն անցան։")
```

## Վարժություն 8․ Նույն փոխարկումը կատարեք երկու ձևով

Վերցրեք առաջին հինգ պսակաթերթերի երկարությունները։ Դրանք միլիմետրի փոխարկեք նախ `for` ցիկլով, ապա NumPy-ի մեկ արտահայտությամբ։ Ցիկլի արդյունքը պահեք `loop_mm` ցուցակում, իսկ NumPy-ի արդյունքը՝ `vector_mm` զանգվածում։

**Հուշում։** Ցիկլում ամեն թիվ բազմապատկեք 10-ով և ավելացրեք ցուցակին՝ `.append()`-ով։

Բջիջների հերթականությունն է՝ Ձեր լուծումը, ստուգումները, առաջարկվող լուծումը։

```python
# Ձեր լուծումը
values_cm = X[:5, 2]
loop_mm = []
# Այստեղ գրեք for ցիկլը։
for value in values_cm:
    loop_mm.append(value * 10)

vector_mm = values_cm * 10
```

```python
# Ստուգումներ
assert len(loop_mm) == 5
assert isinstance(vector_mm, np.ndarray)
assert vector_mm.shape == (5,)
assert np.allclose(loop_mm, [14, 14, 13, 15, 14])
assert np.allclose(loop_mm, vector_mm)
print("Ստուգումներն անցան։")
```

```python
# Առաջարկվող լուծումը․ նախ փորձեք ինքնուրույն։
values_cm = X[:5, 2]
loop_mm = []
for value in values_cm:
    loop_mm.append(value * 10)

vector_mm = values_cm * 10
print("Ցիկլով՝", loop_mm)
print("NumPy-ով՝", vector_mm)
```

## Վարժություն 9․ Ընտրեք երկար պսակաթերթերով ծաղիկները

Ընտրեք այն ծաղիկները, որոնց պսակաթերթի երկարությունը **խիստ մեծ է 5 սմ-ից**։ Պայմանը պահեք `long_mask`, ընտրված տողերը՝ `long_flowers`, իսկ նրանց քանակը՝ `long_count` փոփոխականներում։

**Հուշում։** Նախ գրեք մեկ համեմատություն, հետո այդ դիմակով ընտրեք `X`-ի տողերը։ Քանակի համար օգտագործեք ընտրված աղյուսակի `.shape[0]`-ը։

Բջիջների հերթականությունն է՝ Ձեր լուծումը, ստուգումները, առաջարկվող լուծումը։

```python
# Ձեր լուծումը
long_mask = X[:, 2] > 5.0
long_flowers = X[long_mask]
long_count = long_flowers.shape[0]
```

```python
# Ստուգումներ
assert long_mask.shape == (150,)
assert long_mask.dtype == np.bool_
assert np.array_equal(long_mask, X[:, 2] > 5.0)
assert long_flowers.shape == (42, 4)
assert np.allclose(long_flowers, X[long_mask])
assert long_count == 42
print("Ստուգումներն անցան։")
```

```python
# Առաջարկվող լուծումը․ նախ փորձեք ինքնուրույն։
long_mask = X[:, 2] > 5.0
long_flowers = X[long_mask]
long_count = long_flowers.shape[0]
print("5 սմ-ից մեծ պսակաթերթի երկարությամբ ծաղիկներ՝", long_count)
```

## Վարժություն 10․ Ստացեք չորս առանձին միջինները

Յուրաքանչյուր սյունակի միջինը հաշվեք մեկ արտահայտությամբ և պահեք `column_means` փոփոխականում։ Այնուհետև տպեք երրորդ արժեքը և ասեք՝ որ չափման միջինն է դա։

**Հուշում։** Օգտագործեք `X.mean(axis=0)`։ Երրորդ արժեքի ինդեքսը 2 է։

Բջիջների հերթականությունն է՝ Ձեր լուծումը, ստուգումները, առաջարկվող լուծումը։

```python
# Ձեր լուծումը
column_means = X.mean(axis=0)
print(column_means)
# Այստեղ տպեք երրորդ արժեքը։
print(column_means[2])
```

```python
# Ստուգումներ
assert column_means.shape == (4,)
assert np.allclose(column_means, [5.8433333333, 3.0573333333, 3.758, 1.1993333333])
assert np.isclose(column_means[2], X[:, 2].mean())
print("Ստուգումներն անցան։")
```

```python
# Առաջարկվող լուծումը․ նախ փորձեք ինքնուրույն։
column_means = X.mean(axis=0)
print("Չորս չափումների միջինները՝", column_means)
print("Պսակաթերթի երկարության միջինը՝", column_means[2], "սմ")
```

## Միասին ձևակերպենք արդյունքը

Վարժությունները լուծելուց հետո թվերը ներկայացնենք պարզ նախադասություններով։

- Ուսումնասիրեցինք 150 ծաղկի չորս չափում։

- 42 ծաղկի պսակաթերթի երկարությունը խիստ մեծ է 5 սմ-ից։

Այս եզրակացությունները վերաբերում են մեր տվյալներին։ Դրանք չեն նշանակում, որ բնության բոլոր ծաղիկներն ունեն նույն չափումները։

### Կարճ ամփոփում

1. Ի՞նչ է ընտրում `X[0]`-ը։ **Առաջին ծաղկի բոլոր չափումները։**
2. Ի՞նչ է ընտրում `X[:, 2]`-ը։ **Բոլոր ծաղիկների պսակաթերթերի երկարությունները։**

Եթե կարողանում եք այս երեք հարցին պատասխանել և բացատրել Ձեր հաշվարկների արդյունքները, դասի հիմնական նպատակը կատարված է։
