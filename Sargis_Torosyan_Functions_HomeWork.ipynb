Առաջադրանք 1 Հաճախորդի անվան մաքրում
Հեռացրեք եզրային բացատները, ներքին կրկնվող բացատները դարձրեք մեկ բացատ և
յուրաքանչյուր բառի առաջին տառը դարձրեք մեծատառ՝ .title()-ով։ Վերադարձրեք մաքրված տողը։
Միայն բացատների դեպքում արդյունքը դատարկ տող է։

```python
def clean_customer_name(text):
    return text.strip().replace('  ', ' ').title()
```

```python
assert clean_customer_name(' ani sargsyan ') == 'Ani Sargsyan'
assert clean_customer_name('ARAM') == 'Aram'
assert clean_customer_name(' ') == ''
print('ok')
```

Առաջադրանք 2 Գործարքի տողի գումար
Քանակը վերջավոր թիվ է և կարող է բացասական լինել։ Միավորի գինը ոչ բացասական վերջավոր
թիվ է։ Վերադարձրեք դրանց արտադրյալը՝ կլորացված երկու տասնորդական նիշով։ Պահպանեք
բացասական քանակի նշանը։ Արդյունքը պետք է վերադարձվի, ոչ միայն տպվի։

```python
def calculate_line_total(quantity, unit_price):
    return round(quantity * unit_price, 2)
```

```python
assert calculate_line_total(6, 2.55) == 15.3
assert calculate_line_total(-12, 1.65) == -19.8
assert calculate_line_total(0, 5) == 0
print('ok')
```

Առաջադրանք 3 Զեղչ և լռելյայն արժեք
Երկու մուտքերն էլ վերջավոր թվեր են։ Եթե գումարը բացասական է կամ զեղչի բաժինը 0-ից 1
միջակայքից դուրս է, վերադարձրեք None։ Մնացած դեպքերում վերադարձեք amount × (1 − rate)
արդյունքը՝ երկու նիշով։ Լռելյայն զեղչը զրո է։ 1 բաժինը նշանակում է ամբողջ գումարի զեղչ։
Լրացուցիչ փորձեք բացասական գումարը և rate=1 սահմանը։

```python
def apply_discount(amount, rate=0.0):
    if amount < 0 or (rate < 0 or rate > 1):
        return None

    return round(amount * (1 - rate), 2) 
```

```python
assert apply_discount(20000) == 20000.0
assert apply_discount(20000, rate=0.10) == 18000.0
assert apply_discount(20000, 1.2) is None
print('ok')
```

Առաջադրանք 4 Պահեստի վիճակի որոշում
Պաշարը ամբողջ թիվ կամ None է, իսկ low_limit-ը՝ դրական ամբողջ թիվ։ None → «Բացակայում
է», բացասական թիվ → «Սխալ», 0 → «Սպառված», 1-ից մինչև low_limit ներառյալ → «Քիչ»,
դրանից մեծ → «Բավարար»։ Վերադարձրեք ճիշտ մեկ պիտակ։
Լրացուցիչ փորձեք 0, -1 և low_limit+1 արժեքները։

```python
def stock_label(stock, low_limit=5):
    if stock is None:
        return 'Բացակայում է'
    
    if stock < 0:
        return 'Սխալ'

    if stock == 0:
        return 'Սպառված'

    if 1 <= stock <= low_limit:
        return 'Քիչ'

    return 'Բավարար'
```

```python
assert stock_label(None) == 'Բացակայում է'
assert stock_label(5) == 'Քիչ'
assert stock_label(5, low_limit=3) == 'Բավարար'
print('ok')
```

Առաջադրանք 5 Միջին արժեքի անվտանգ հաշվարկ
Մուտքը վերջավոր թվերի ցուցակ է։ Ցիկլով հաշվեք գումարը և տարրերի քանակը։ Դատարկ
ցուցակի դեպքում վերադարձրեք None, հակառակ դեպքում՝ միջինը երկու նիշով։ Կուտակիչները
ստեղծեք ֆունկցիաի ներսում։ Պատրաստ sum() մի օգտագործեք։

```python
def safe_average(values):
    values_cnt = len(values)

    if values_cnt == 0:
        return None
    
    total = 0

    for value in values:
        total += value

    return round(total / values_cnt, 2)
```

```python
assert safe_average([10, 20, 30]) == 20.0
assert safe_average([]) is None
assert safe_average([0, 0]) == 0.0
print('ok')
```

Առաջադրանք 6 Նկարագրությունների ցուցակի մաքրում
Ցուցակի տարրերը տողեր կամ None են։ Վերադարձրեք նոր ցուցակ՝ նույն հերթականությամբ։
Տողերի կրկնվող բացատները միավորեք և տառերը դարձրեք մեծատառ։ None-ը պահեք None,
միայն բացատներից կազմված տողը՝ դատարկ տող։ Սկզբնական ցուցակը մի փոփոխեք։
Նույն մուտքով կրկնվող կանչերը պետք է տան անկախ նոր ցուցակներ։

```python
def clean_descriptions(values):
    new_values = []

    for value in values:
        if value is not None:
            value = ' '.join(value.split()).upper()

        new_values.append(value)

    return new_values
```

```python
assert clean_descriptions([' a b ', None]) == ['A B', None]
assert clean_descriptions([' ']) == ['']
assert clean_descriptions([]) == []
print('ok')
```

Առաջադրանք 7 Շեմին հասած արժեքների քանակ
Մուտքը վերջավոր թվերի ցուցակ է, շեմը՝ թիվ։ Վերադարձրեք շեմին հավասար կամ դրանից մեծ
արժեքների քանակը։ Հաշվիչը ստեղծեք ֆունկցիաի ներսում։ Դատարկ ցուցակի համար
վերադարձրեք 0։

```python
def count_at_least(values, threshold=100):
    count = 0

    for value in values:
        if value >= threshold:
            count += 1

    return count
```

```python
assert count_at_least([99, 100, 120]) == 2
assert count_at_least([2, 5, 8], threshold=5) == 2
assert count_at_least([]) == 0
print('ok')
```

Առաջադրանք 8 Առաջին համապատասխան ծածկագիրը
Ծածկագրերը տողերի ցուցակ են։ Նախածանցը մաքրումից հետո ոչ դատարկ տող է։ Եզրային
բացատները հեռացրեք և մեծատառ դարձրեք թե՛ նախածանցը, թե՛ ստուգվող ծածկագիրը։
Վերադարձրեք առաջին համապատասխան մաքրված ծածկագիրը։ Չգտնվելու դեպքում
վերադարձրեք None։
Դատարկ ցուցակի դեպքում նույնպես պետք է ստացվի None։

```python
def first_matching_code(codes, prefix="C"):
    result = None
    prefix_upper = prefix.upper().strip()

    for code in codes:
        code_upper = code.upper().strip()

        if code_upper.startswith(prefix_upper):
            result = code_upper
            break
    
    return result
```

```python
assert first_matching_code(['A1', ' c20 ', 'C30']) == 'C20'
assert first_matching_code([' a1 ', 'B2'], prefix='a') == 'A1'
assert first_matching_code(['A1', 'B2']) is None
print('ok')
```

Առաջադրանք 9 Հաշվի համարի պիտակ
Մուտքը տող է։ Հեռացրեք եզրային բացատները և մեծատառ դարձրեք տառերը։ Դատարկ
արդյունքի համար վերադարձեք «Դատարկ համար»։ C-ով սկսվող համարի համար՝ «Չեղարկում»։
Մնացածի համար՝ «Չեղարկման նշում չկա»։ Սա միայն նշման ճանաչում է, ոչ համարի
ամբողջական վավերացում։

```python
def invoice_kind(invoice):
    stripped_invoice = invoice.strip()

    if stripped_invoice == '':
        return 'Դատարկ համար'

    stripped_invoice = stripped_invoice.upper()

    if stripped_invoice.startswith('C'):
        return 'Չեղարկում'

    return 'Չեղարկման նշում չկա'
```

```python
assert invoice_kind(' c123 ') == 'Չեղարկում'
assert invoice_kind('536365') == 'Չեղարկման նշում չկա'
assert invoice_kind(' ') == 'Դատարկ համար'
print('ok')
```

Առաջադրանք 10 Քանակների ամփոփում բառարանով
Մուտքը ամբողջ թվերի ցուցակ է։ Վերադարձրեք ճիշտ երեք բանալիով բառարան՝ positive՝
դրականների քանակը, negative՝ բացասականների քանակը, zero՝ զրոների քանակը։
Յուրաքանչյուր տարրը պետք է հաշվվի միայն մեկ խմբում։ Դատարկ մուտքի բոլոր հաշվիչները 0
են։

```python
def summarize_quantities(values):
    quantities = {'positive': 0, 'negative': 0, 'zero': 0}

    for value in values:
        if value == 0:
            quantities['zero'] += 1
            continue

        if value > 0:
            quantities['positive'] += 1
            continue

        quantities['negative'] += 1

    return quantities
```

```python
assert summarize_quantities([2, -1, 0, 5]) == {
 'positive': 2, 'negative': 1, 'zero': 1
}
assert summarize_quantities([]) == {
 'positive': 0, 'negative': 0, 'zero': 0
}
assert summarize_quantities([0, 0]) == {
 'positive': 0, 'negative': 0, 'zero': 2
}
print('ok')
```

Առաջադրանք 11 Մաքրված գրառում՝ սկզբնականը
պահպանելով
Բառարանում Description դաշտը տող կամ None է, իսկ մյուս արժեքները պարզ տողեր, թվեր կամ
None են։ Վերադարձրեք նոր բառարան՝ բոլոր դաշտերը պահպանելով։ Փոխեք միայն
Description-ը՝ միավորված բացատներով և մեծատառերով։ None-ը պահպանեք։ Սկզբնական
բառարանը մի փոփոխեք։
Լրացուցիչ ստուգեք, որ վերադարձված բառարանը սկզբնական նույն օբյեկտը չէ։

```python
row = {'Description': ' a b ', 'Quantity': 2}

def make_clean_record(record):
    new_record = record.copy()
    description = new_record['Description']

    if description is not None:
        new_record['Description'] = ' '.join(description.split()).upper()

    return new_record
```

```python
cleaned = make_clean_record(row)
assert cleaned == {'Description': 'A B', 'Quantity': 2}
assert row == {'Description': ' a b ', 'Quantity': 2}
assert make_clean_record({'Description': None}) == {
 'Description': None
}
print('ok')
```

Առաջադրանք 12 Վաճառքի ընդունելի տողի ստուգում
Բոլոր դաշտերը առկա են։ Description-ը տող կամ None է, InvoiceNo-ն տող, քանակն ու գինը՝
վերջավոր թվեր։ Վերադարձրեք True միայն եթե նկարագրությունը առկա և մաքրումից հետո ոչ
դատարկ է, հաշվի համարը C-ով չի սկսվում, քանակն ու գինը դրական են։ Մնացած դեպքերում՝
False։ Հաշվի համարն ստուգելուց առաջ հեռացրեք բացատները և մեծատառ դարձրեք։
Լրացուցիչ փորձեք զրոյական գին և միայն բացատներից կազմված նկարագրություն։

```python
sale = {
 'Description': 'Գրիչ', 'InvoiceNo': '100',
 'Quantity': 2, 'UnitPrice': 10.5, 'Country': 'AM'
}
cancelled = sale.copy()
cancelled['InvoiceNo'] = 'C100'
cancelled['Quantity'] = -2

def is_sale_record(row):
    cleaned_invoice_no = row['InvoiceNo'].strip().upper()
    description = row['Description']

    return description is not None and description.strip() != '' and not cleaned_invoice_no.startswith('C') and row['Quantity'] > 0 and row['UnitPrice'] > 0
```

```python
assert is_sale_record(sale) is True
assert is_sale_record(cancelled) is False
missing = sale.copy()
missing['Description'] = None
assert is_sale_record(missing) is False
print('ok')
```

Առաջադրանք 13 Ընդունված վաճառքների ամփոփում
Օգտագործեք calculate_line_total() և is_sale_record() ֆունկցիաները։ Վերադարձրեք ճիշտ երեք
բանալիով բառարան՝ count՝ ընդունված տողերի քանակ, total՝ դրանց գումար, average՝ միջին
գումար։ Գումարն ու միջինը կլորացրեք երկու նիշով։ Եթե ընդունված տող չկա, count-ը և total-ը 0
են, average-ը՝ None։ Միջինը հաշվեք ընդունվածների քանակով։
Այս առաջադրանքը կախված է 2-րդ և 12-րդ առաջադրանքների ֆունկցիաներից։

```python
sale = {
 'Description': 'Գրիչ', 'InvoiceNo': '100',
 'Quantity': 2, 'UnitPrice': 10.5, 'Country': 'AM'
}
cancelled = sale.copy()
cancelled['InvoiceNo'] = 'C100'
cancelled['Quantity'] = -2

def summarize_sales(rows):
    sales = {'count': 0, 'total': 0, 'average': None}

    for row in rows:

        if is_sale_record(row):
            sales['total'] += calculate_line_total(row['Quantity'], row['UnitPrice'])
            sales['count'] += 1

    if sales['count'] != 0:
        sales['average'] = round(sales['total'] / sales['count'], 2)

    return sales
```

```python
assert summarize_sales([sale, cancelled]) == {
 'count': 1, 'total': 21.0, 'average': 21.0
}
assert summarize_sales([]) == {
 'count': 0, 'total': 0, 'average': None
}
assert summarize_sales([cancelled]) == {
 'count': 0, 'total': 0, 'average': None
}
print('ok')
```

Առաջադրանք 14 Վաճառքի գումարներն ըստ երկրի
Օգտագործեք նույն երկու ֆունկցիաները՝ calculate_line_total() և is_sale_record()։ Վերադարձրեք
բառարան, որի բանալիները երկրներն են, արժեքները՝ այդ երկրի ընդունված տողերի
գումարները՝ երկու նիշով։ Առանց ընդունված տողի երկիրը մի ներառեք։ Դատարկ մուտքից
վերադարձրեք դատարկ բառարան։
Կախված է 2-րդ և 12-րդ առաջադրանքների ֆունկցիաներից։ Նույն երկրի կրկնվող տողերը պետք
է գումարվեն։

```python
sale = {
 'Description': 'Գրիչ', 'InvoiceNo': '100',
 'Quantity': 2, 'UnitPrice': 10.5, 'Country': 'AM'
}
cancelled = sale.copy()
cancelled['InvoiceNo'] = 'C100'
cancelled['Quantity'] = -2
second = sale.copy()
second['Quantity'] = 4
second['UnitPrice'] = 5
second['Country'] = 'GE'

def group_sales_by_country(rows):
    sales_by_country = {}

    for row in rows:
        if is_sale_record(row):
            line_total = calculate_line_total(row['Quantity'], row['UnitPrice'])
            country = row['Country']

            if country not in sales_by_country:
                sales_by_country[country] = line_total
            else:
                sales_by_country[country] = round(sales_by_country[country] + line_total, 2)

    return sales_by_country    
```

```python
assert group_sales_by_country([sale, second]) == {
 'AM': 21.0, 'GE': 20.0
}
assert group_sales_by_country([sale, sale, cancelled]) == {
 'AM': 42.0
}
assert group_sales_by_country([]) == {}
print('ok')
```

Առաջադրանք 15 Ամբողջական փոքր վերլուծություն
Կազմեք վերջնական հաշվետվություն՝ օգտագործելով make_clean_record(), summarize_sales() և
group_sales_by_country() ֆունկցիաները։ Նախ ստեղծեք բոլոր գրառումների մաքրված նոր
ցուցակը։ Վերադարձրեք ճիշտ երեք բանալիով բառարան՝ rows՝ մաքրված բոլոր գրառումները՝
սկզբնական հերթականությամբ, summary՝ դրանց վաճառքային ամփոփումը, by_country՝
ընդունված վաճառքների գումարներն ըստ երկրի։ Սկզբնական ցուցակն ու բառարանները մի
փոփոխեք։
Կախված է 11-րդ, 13-րդ և 14-րդ առաջադրանքներից։ Երեք տողով հայերեն բացատրեք՝ որ
ֆունկցիան ինչ պատասխանատվություն ունի։

```python
sale = {
 'Description': ' pen ', 'InvoiceNo': '100',
 'Quantity': 2, 'UnitPrice': 10.5, 'Country': 'AM'
}
rows = [sale]

def build_report(rows):
    report = {
       'rows': [],
       'summary': summarize_sales(rows),
       'by_country': group_sales_by_country(rows) 
    }

    for row in rows:
        report['rows'].append(make_clean_record(row))

    return report
```

```python
report = build_report(rows)
assert report == {
 'rows': [dict(sale, Description='PEN')],
 'summary': {'count': 1, 'total': 21.0, 'average': 21.0},
 'by_country': {'AM': 21.0}
}
assert sale['Description'] == ' pen '
assert build_report([]) == {
 'rows': [],
 'summary': {'count': 0, 'total': 0, 'average': None},
 'by_country': {}
}
print('ok')
```
