## 2. Փոխարկում և ստուգում
Միասին գրեք `parse_row(row)`։ Վերադարձրեք նոր բառարան՝ `alcohol` և `pH` դաշտերը դարձնելով `float`, իսկ `quality`-ն՝ `int`։ Եթե թվի փոխարկումը չի հաջողվում, թողեք, որ բարձրանա `ValueError`։ Փորձարկման սխալ տողը հորինված է։

Հետո գրեք `prepare_rows(raw_rows)`, որը կիրառում է `parse_row()`-ը բոլոր տողերի վրա և վերադարձնում նոր ցուցակ։ Քննարկեք, թե որ փոփոխականներն են հասանելի միայն ֆունկցիայի ներսում։


```python
def parse_row(row):
    return {
        "alcohol": float(row["alcohol"]),
        "pH": float(row["pH"]),
        "quality": int(row["quality"]),
    }


def prepare_rows(raw_rows):
    return [parse_row(row) for row in raw_rows]
```

```python
records = prepare_rows(raw_rows)
assert len(records) == 1599
assert records[0] == {"alcohol": 9.4, "pH": 3.51, "quality": 5}
assert raw_rows[0]["quality"] == "5"
assert prepare_rows([]) == []

try:
    parse_row({"alcohol": "?", "pH": "3.51", "quality": "5"})
except ValueError:
    print("Սխալ թիվը հայտնաբերվեց։")
else:
    raise AssertionError("Սպասվում էր ValueError։")

print('passed')
```

### Վարժություն 2 · Դաշտի փոխարկում

Տողը փոխարկեք float-ի։ Եթե տողը թվի չի փոխարկվում, վերադարձրեք default արժեքը։

```python
def read_number(text, default=None):
    try:
        return float(text)
    except:
        return default
```

```python
assert read_number(' 3.51 ') == 3.51
assert read_number('?', default=0.0) == 0.0
print('passed')
```

## 3. Ֆիլտրում և լռելյայն արժեք
Գրեք `filter_by_quality(rows, minimum=7)`։ Ընտրեք բոլոր տողերը, որոնց գնահատականը մեծ կամ հավասար է շեմին։ Ստուգեք լռելյայն կանչը, անվանված արգումենտով կանչը և դատարկ ցուցակը։ Նախ կանխատեսեք, թե ինչ կլինի հենց 7 գնահատականով տողի հետ։


```python
def filter_by_quality(rows, minimum=7):
    return [row for row in rows if row["quality"] >= minimum]
```

```python
assert filter_by_quality([]) == []
assert filter_by_quality([{"quality": 6}, {"quality": 7}]) == [{"quality": 7}]
assert len(filter_by_quality(records)) == 217
assert len(filter_by_quality(records, minimum=8)) == 18
assert filter_by_quality(records, minimum=9) == []
print('passed')
```

### Բուլյան արդյունք և տեղային փոփոխականներ
Պայմանի արդյունքը կարող է լինել `True` կամ `False`։ `quality`-ն տեղային փոփոխական է։ Շեմը փոխանցում ենք որպես պարամետր, որպեսզի ֆունկցիան աշխատի տարբեր շեմերով։

```python
def meets_quality(row, minimum=7):
    quality = row["quality"]
    return quality >= minimum

print(meets_quality(records[0], minimum=5))
```

### Վարժություն 3 · Շեմին հասած դիտարկումների քանակ

Հաշվեք minimum կամ ավելի բարձր quality ունեցող տողերը։ Դատարկ ցուցակի դեպքում վերադարձրեք 0։

```python
def count_at_least(rows, minimum=7):
    return len(filter_by_quality(rows, minimum))
    
```

```python
assert count_at_least([{'quality': 6}, {'quality': 7}]) == 1
assert count_at_least(records) == 217
assert count_at_least([], minimum=5) == 0
print('passed')
```

### Վարժություն 4 · Փակ միջակայքով ֆիլտրում

Վերադարձրեք այն տողերը, որոնց alcohol արժեքը low և high սահմանների միջև է՝ ներառյալ երկու սահմանները։ Ենթադրեք low ≤ high։

```python
def select_alcohol_range(rows, low, high):
    return [row for row in rows if low <= row['alcohol'] <= high]
```

```python
assert select_alcohol_range([{'alcohol': 9.0}, {'alcohol': 10.0}, {'alcohol': 11.0}], 9.0, 10.0) == [{'alcohol': 9.0}, {'alcohol': 10.0}]
print('passed')
```

## 4. Խմբավորում և հաշվարկ
Գրեք `group_by_quality(rows)`, որը վերադարձնում է բառարան՝ գնահատականից դեպի այդ գնահատականով տողերի ցուցակ։ Ապա գրեք `mean_value(rows, field)`, որը հաշվում է թվային դաշտի միջինը։ Դատարկ ցուցակի համար վերադարձրեք `None`։

Միևնույն ֆունկցիայով հաշվեք թե՛ `alcohol`-ի, թե՛ `pH`-ի միջինները։ Կլորացրեք ցուցադրելու պահին։


```python
from statistics import mean


def group_by_quality(rows):
    groups = {}
    for row in rows:
        groups.setdefault(row["quality"], []).append(row)
    return groups


def mean_value(rows, field):
    if not rows:
        return None
    return mean(row[field] for row in rows)
```

```python
groups = group_by_quality(records)
assert sum(len(group) for group in groups.values()) == 1599
assert len(groups[5]) == 681
assert group_by_quality([]) == {}
assert mean_value([], "pH") is None
assert mean_value([{"pH": 3.0}, {"pH": 4.0}], "pH") == 3.5
print('passed')
```

### Վարժություն 5 · Գնահատականների հաճախականություն

Վերադարձրեք բառարան, որի բանալին quality-ն է, իսկ արժեքը՝ այդ գնահատականով տողերի թիվը։

```python
def quality_counts(rows):
    quality_count = {}
    for row in rows:
        if row["quality"] not in quality_count:
            quality_count[row["quality"]] = 1
        else:
            quality_count[row["quality"]] += 1

    return quality_count
```

```python
assert quality_counts([]) == {}
assert quality_counts(records)[5] == 681
assert sum(quality_counts(records).values()) == 1599
print('passed')
```

### Երկու արդյունքի վերադարձ
`return low, high`-ը վերադարձնում է tuple։ Այն կարելի է բացել երկու փոփոխականի մեջ։ Դատարկ մուտքը ստուգեք մինչև `min()` կամ `max()` կանչելը։

### Վարժություն 6 · Նվազագույն և առավելագույն

Տրված թվային դաշտի համար վերադարձրեք (նվազագույն, առավելագույն) զույգը։ Դատարկ ցուցակի դեպքում վերադարձրեք None։

```python
def field_bounds(rows, field):
    if not rows:
        return None
    
    min_value = None
    max_value = None

    for row in rows:
        if min_value is None or row[field] < min_value:
            min_value = row[field]

        if max_value is None or row[field] > max_value:
            max_value = row[field]

    return (min_value, max_value)
```

```python
assert field_bounds([{'pH': 3.5}, {'pH': 3.0}], 'pH') == (3.0, 3.5)
assert field_bounds([], 'alcohol') is None
print('passed')
```

```python
low, high = field_bounds(records, "alcohol")
print(low, high)
```

## 5. Խնդրի բաժանում
Երբ ֆունկցիան արդեն հաշվում է ընտրված տողերի քանակը, բաժինը հաշվելիս կարող ենք օգտագործել նույն ֆունկցիան։ Այսպես շեմի կանոնը գրված է մեկ տեղում։

### Վարժություն 7 · Շեմը բավարարող տողերի բաժին

Օգտագործեք count_at_least() ֆունկցիան։ Արդյունքը վերադարձրեք որպես 0-ից 1 թիվ։ Դատարկ ցուցակի դեպքում վերադարձրեք None։

```python
def quality_share(rows, minimum=7):
    if not rows:
        return None
    
    count = count_at_least(rows, minimum)

    return count / len(rows)
```

```python
assert quality_share(records) == 217 / 1599
assert quality_share(records, minimum=9) == 0
assert quality_share([]) is None
print('passed')
```

### Վարժություն 8 · Երկու խմբի միջինների տարբերություն

Օգտագործեք դասի group_by_quality() և mean_value() ֆունկցիաները։ Առաջին գնահատականի խմբի միջինից հանեք երկրորդ խմբի միջինը։ Եթե որևէ խումբ չկա, վերադարձրեք None։

```python
def compare_means(rows, first, second, field="alcohol"):
    groups = group_by_quality(rows)

    if first not in groups or second not in groups:
        return None

    first_group = groups[first]
    second_group = groups[second]

    return mean_value(first_group, field) - mean_value(second_group, field)
```

```python
assert round(compare_means(records, 7, 5), 2) == 1.57
assert compare_means(records, 7, 9) is None
print('passed')
```

### Վարժություն 9 · Ամենահաճախ հանդիպող գնահատական

Օգտագործեք quality_counts() ֆունկցիան։ Հավասար հաճախականության դեպքում ընտրեք փոքր գնահատականը։ Դատարկ ցուցակի դեպքում վերադարձրեք None։

```python
def most_common_quality(rows):
    if not rows:
        return None
    
    quality_count = quality_counts(rows)

    max_count = max(quality_count.values())
    most_common_qualities = [quality for quality, count in quality_count.items() if count == max_count]

    return min(most_common_qualities)
```

```python
#assert most_common_quality(records) == 5
assert most_common_quality([{'quality': 7}, {'quality': 5}]) == 5
assert most_common_quality([]) is None
print('passed')
```

### Վարժություն 10 · Փոքր վերլուծական հաշվետվություն

Կիրառեք նախկինում գրված ֆունկցիաները և վերադարձրեք բառարան՝ total, selected, share, mean_ph բանալիներով։ mean_ph-ը հաշվեք միայն շեմը բավարարող տողերի համար։ Եթե այդպիսի տող չկա, mean_ph-ը None է։

```python
def threshold_report(rows, minimum=7):
    selected_rows = filter_by_quality(rows, minimum)
    mean_ph = mean_value(selected_rows, "pH")

    result = {
        "total": len(rows),
        "selected": len(selected_rows),
        "share": None,
        "mean_ph": mean_ph,
    }

    if result["total"] > 0:
        result["share"] = result["selected"] / result["total"]

    return result
```

```python
assert threshold_report([{'quality': 6, 'pH': 3.2}, {'quality': 7, 'pH': 3.4}]) == {'total': 2, 'selected': 1, 'share': 0.5, 'mean_ph': 3.4}
assert threshold_report([], minimum=8) == {'total': 0, 'selected': 0, 'share': None, 'mean_ph': None}
assert threshold_report(records, minimum=8)['selected'] == 18
print('passed')
```

## 5. Վերջնական վերլուծություն զույգերով
Առաջին մասնակիցը գրում է `build_report(rows, minimum=None)`։ Եթե շեմը տրված է, կիրառեք ֆիլտրը։ Խմբավորեք տվյալները և վերադարձրեք գնահատականով աճման կարգավորված ցուցակ՝ `quality`, `count`, `mean_alcohol` և `mean_ph` դաշտերով։ Երկրորդ մասնակիցը գրում է ստուգումները։ Հետո փոխեք դերերը և միասին կառուցեք ընթեռնելի աղյուսակը։

**Հարց։** Համեմատեք 5 և 7 գնահատականներով խմբերը։ Նշեք երկու խմբերի քանակները և միջին ալկոհոլի տարբերությունը տոկոսային կետերով։ Բացատրեք, թե ինչու 8 գնահատականով փոքր խումբը զգայուն է առանձին չափումների նկատմամբ։


```python
def build_report(rows, minimum=None):
    selected = rows if minimum is None else filter_by_quality(rows, minimum)
    groups = group_by_quality(selected)
    report = []
    for quality, group in sorted(groups.items()):
        report.append({
            "quality": quality,
            "count": len(group),
            "mean_alcohol": mean_value(group, "alcohol"),
            "mean_ph": mean_value(group, "pH"),
        })
    return report
```

```python
report = build_report(records)
assert [row["quality"] for row in report] == [3, 4, 5, 6, 7, 8]
assert sum(row["count"] for row in build_report(records, minimum=7)) == 217
assert build_report(records, minimum=9) == []
print("Որակ | Քանակ | Միջին ալկոհոլ (%) | Միջին pH")
for row in report:
    print(f"{row['quality']:>4} | {row['count']:>5} | "
          f"{row['mean_alcohol']:>17.2f} | {row['mean_ph']:>8.2f}")

by_quality = {row["quality"]: row for row in report}
difference = by_quality[7]["mean_alcohol"] - by_quality[5]["mean_alcohol"]
print(f"Տարբերությունը {difference:.2f} տոկոսային կետ է։")
```

### Ծրագրի մուտքի կետ
Հաշվարկող ֆունկցիաները վերադարձնում են տվյալներ։ `main()`-ը կապում է փոխարկումը, հաշվարկը և ցուցադրումը։ Փոխեք շեմը ու համեմատեք արդյունքները։

```python
def main(raw_rows, minimum=7):
    rows = prepare_rows(raw_rows)
    summary = threshold_report(rows, minimum)
    print("Ընդհանուր", summary["total"])
    print("Ընտրված", summary["selected"])
    if summary["share"] is not None:
        print(f"Ընտրվածների բաժին {summary['share']:.1%}")
    return summary

summary_7 = main(raw_rows)
summary_8 = main(raw_rows, minimum=8)
assert summary_7["selected"] == 217
assert summary_8["selected"] == 18
```

Վերջում համեմատեք 7 և 8 շեմերով հաշվետվությունները։ Նշեք, թե դիտարկումների ինչ բաժին է մնում յուրաքանչյուր շեմի դեպքում։
