# Պայմանական կառույցների գործնական առաջադրանքներ

**Ռաֆիկ Առաքելյան · ԳործԱփ ակադեմիա**

Լուծեք առաջադրանքները հերթով։ Պահանջվում է Python 3.10 կամ ավելի նոր։ Տվյալները ներառված են հաջորդ կոդային բջիջում։ Առաջադրանքները նախատեսված են դասարանում ինքնուրույն կամ զույգերով աշխատանքի համար։

Յուրաքանչյուր առաջադրանքի տրված կոդը պատճենեք լուծման բջիջի մեջ և շարունակեք այն։ Ստեղծեք պահանջված անուններով փոփոխականները։ Արդյունքները հաշվեք մուտքերից. սպասվող պատասխանները որպես պատրաստ արժեք մի գրեք։

Ստուգումների բջիջը գործարկեք լուծումից հետո։ Յուրաքանչյուր առաջադրանք ունի երեք `assert` ստուգում։ Դատարկ լուծումների դեպքում սխալը սպասելի է։ Ստուգվում են միայն սկզբնական մուտքերի արդյունքները. բոլոր ճյուղերի ճիշտ աշխատանքը դրանցով չի ապացուցվում։

Լրացուցիչ փորձերի ժամանակ վերականգնեք սկզբնական մուտքերը, հետո փոխեք միայն նշված արժեքները։ Այդ փորձերի վրա սկզբնական ստուգումները մի գործարկեք։ Ավարտելուց հետո վերադարձեք սկզբնական մուտքերին, վերագործարկեք միջավայրը և գործարկեք ամբողջ ֆայլը։

Սկզբնական տվյալները մի փոփոխեք։ Հատուկ փոփոխված օրինակ ստեղծելիս օգտագործեք բառարանի `.copy()` պատճենը։ Ցիկլեր կամ սեփական գործառույթներ պետք չեն։

## Առաջադրանք 3 · Գումարի խմբավորում

Հաշվեք `value` գումարը։ Այս առաջադրանքում ընտրված տողը դրական քանակ և գին ունի։ Ստեղծեք `value_band`՝ առնվազն 100՝ «Մեծ», առնվազն 50 և 100-ից փոքր՝ «Միջին», մնացածը՝ «Փոքր»։ Ստեղծեք `needs_attention`՝ ճիշտ է միայն «Մեծ» խմբի համար։ Օգտագործեք մեկ պայմանական շղթա։

### Տրված է

```python
row = by_row[34]
```

### Լրացուցիչ փորձ

Տողը փոխեք 247-ով. սպասվում է 165.00, «Մեծ», `True`։ Հետո հաշվարկի փոխարեն առանձին փորձեք `value = 50` և `value = 100`։

````python
# Գրեք այստեղ լուծումը
row = by_row[34]
value = round(row['Quantity'] * row['UnitPrice'], 2)

needs_attention = False
if value >= 100:
    value_band = 'Մեծ'
    needs_attention = True
elif value >= 50:
    value_band = 'Միջին'
else:
    value_band = 'Փոքր'
````

## Առաջադրանք 4 · Նկարագրության անվտանգ մաքրում

Եթե նկարագրությունը `None` է, թող `cleaned`-ը լինի `None`, իսկ `text_state`-ը՝ «Բացակայում է»։ Հակառակ դեպքում միավորեք կրկնվող բացատները. դատարկ արդյունքի պիտակը «Դատարկ է» է, մյուսինը՝ «Առկա է»։ Ստեղծեք `usable_text`՝ ճիշտ է միայն «Առկա է» վիճակում։

### Տրված է

```python
row = by_row[623]
```

### Լրացուցիչ փորձ

Տողը փոխեք 88-ով. սպասվում է «JUMBO BAG BAROQUE BLACK WHITE», «Առկա է», `True`։ Առանձին ուսումնական պատճենում փորձեք նաև միայն բացատներից կազմված նկարագրություն։

````python
# Գրեք այստեղ լուծումը
row = by_row[623]

usable_text = False
if row['Description'] is None:
    cleaned = None
    text_state = 'Բացակայում է'
else:
    cleaned = row['Description'].replace('  ', ' ')
    if cleaned == '':
        text_state = 'Դատարկ է'
    else:
        text_state = 'Առկա է'
        usable_text = True
````

````python
assert cleaned is None
assert text_state == 'Բացակայում է'
assert usable_text is False
print('Առաջադրանք 4․ ստուգումներն անցան։')
````

## Առաջադրանք 5 · Միաժամանակ առկա խնդիրներ

Ստեղծեք `missing_text` և `zero_price` տրամաբանական արժեքները։ Առաջինը ճիշտ է, երբ նկարագրությունը `None` կամ մաքրումից հետո դատարկ է։ Երկրորդը ճիշտ է, երբ գինը զրո է։ Անկախ `if` հրամաններով կազմեք `problems` ցուցակը՝ նախ «Նկարագրություն», ապա «Գին» պիտակներով։ `problem_count`-ը ցուցակի երկարությունն է։

### Տրված է

```python
row = by_row[623]
```

### Լրացուցիչ փորձ

Տողը փոխեք 41-ով. երկու դրոշներն էլ `False` են, ցուցակը դատարկ է, քանակը՝ 0։

````python
# Գրեք այստեղ լուծումը
row = by_row[623]

missing_text = row['Description'] is None or row['Description'].strip() == ''
zero_price = row['UnitPrice'] == 0

problems = []

if missing_text:
    problems.append('Նկարագրություն')

if zero_price:
    problems.append('Գին')

problem_count = len(problems)
````

````python
assert missing_text is True and zero_price is True
assert problems == ['Նկարագրություն', 'Գին']
assert problem_count == 2
print('Առաջադրանք 5․ ստուգումներն անցան։')
````

## Առաջադրանք 6 · Համակցված ընտրության կանոն

Ստեղծեք `selected`՝ տողը չեղարկված չէ, քանակը դրական է, գինը դրական է և միաժամանակ կամ երկիրը Ֆրանսիան է ու քանակը առնվազն 20, կամ միավորի գինը առնվազն 100 է։ Օգտագործեք փակագծեր։ `selection_reason`-ը լինի «Ընտրված է» կամ «Չի ընտրվել»։ `selected_amount`-ը հաշվարկված գումարն է միայն ընտրված տողի համար, այլապես `None`։

## Առաջադրանք 7 · Գործողության ընտրություն ըստ վիճակի

Օգտագործեք `match / case`։ «Վաճառք» → «Գումարել», «Չեղարկում» → «Առանձին դիտարկել», «Գնի սխալ» կամ «Զրոյական գնի ստուգում» → «Ստուգել գինը», մնացածը → «Ստուգել տվյալները»։ Արդյունքը պահեք `action`-ում։ `include_sale`-ը ճիշտ է միայն «Վաճառք» վիճակում։ `needs_price_check`-ը ճիշտ է գնի երկու վիճակներում։ Այս մուտքային պիտակը ուսումնական է։

### Տրված է

```python
state = 'Գնի սխալ'
```

### Լրացուցիչ փորձ

Փորձեք «Վաճառք»՝ «Գումարել», `True`, `False`։ Փորձեք «Անհայտ»՝ «Ստուգել տվյալները», `False`, `False`։

````python
# Գրեք այստեղ լուծումը
state = 'Գնի սխալ'

include_sale = False
needs_price_check = False
match state:
    case 'Վաճառք':
        action = 'Գումարել'
        include_sale = True
    case 'Չեղարկում':
        action = 'Առանձին դիտարկել'
    case 'Գնի սխալ' | 'Զրոյական գնի ստուգում':
        action = 'Ստուգել գինը'
        needs_price_check = True
    case _:
        action = 'Ստուգել տվյալները'
````

````python
assert action == 'Ստուգել գինը'
assert include_sale is False
assert needs_price_check is True
print('Առաջադրանք 7․ ստուգումներն անցան։')
````

## Առաջադրանք 8 · Լրացուցիչ պայմանով համապատասխանեցում

Մուտքը թիվ է կամ `None`։ `match / case`-ով ստեղծեք `amount_state`՝ `None` → «Բացակայում է», բացասական → «Բացասական», 0 → «Զրո», առնվազն 100 → «Խոշոր դրական», մյուս դրական արժեքները → «Փոքր դրական»։ Ստեղծեք `is_negative` և `is_known` տրամաբանական արժեքները։ Նախ ստուգեք բացակայող արժեքը։

### Տրված է

```python
amount = round(by_row[142]['Quantity'] * by_row[142]['UnitPrice'], 2)
```

### Լրացուցիչ փորձ

Փորձեք `amount = None`՝ «Բացակայում է», `False`, `False`։ Փորձեք `amount = 100`՝ «Խոշոր դրական», `False`, `True`։

````python
# Գրեք այստեղ լուծումը
row = by_row[142]
amount = round(row['Quantity'] * row['UnitPrice'], 2)

match amount:
    case None:
        amount_state = 'Բացակայում է'
    case _ if amount < 0:
        amount_state = 'Բացասական'
    case 0:
        amount_state = 'Զրո'
    case _ if amount >= 100:
        amount_state = 'Խոշոր դրական'
    case _:
        amount_state = 'Փոքր դրական'

is_negative = amount_state == 'Բացասական'
is_known = amount is not None
````

````python
assert amount_state == 'Բացասական'
assert is_negative is True
assert is_known is True
print('Առաջադրանք 8․ ստուգումներն անցան։')
````

## Առաջադրանք 9 · Գտեք ճյուղերի հերթականության սխալը

Ստորև բերված ուսումնական ծրագիրը սխալ է խմբավորում 165 գումարը։ Բացատրեք պատճառը կոդի մեկնաբանությամբ և լուծման բջիջում գրեք ուղղված տարբերակը։ Ստեղծեք նաև `at_least_100` և `below_50` տրամաբանական արժեքները։

```python
if amount >= 50:
```

| Բանալի | Պահանջ |
