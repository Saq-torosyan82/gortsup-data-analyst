sales = 35000

● target_status՝ 'Reached' կամ 'Below target',

● sales_difference՝ sales − target տարբերությունը։

Երկու inputs-ն էլ nonnegative numbers են։ Հենց target-ի արժեքը համարվում է reached։

stock = 5

**Առաջադրանք 3 Discount ի ճիշտ հերթականություն**

**Տրված է**

order_amount = 60000

Order amount-ը nonnegative number է՝ AMD-ով։ Ընտրեք discount_rate-ը․

● առնվազն 60000 → 0.10,

● առնվազն 30000, բայց 60000-ից պակաս → 0.05,

● մնացածը → 0.0։

Հաշվեք discount_amount և final_amount՝ կլորացնելով 2 decimal places-ի։ Օգտագործեք մեկ if / elif / else chain։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ order_amount = 29999։ Սպասվող՝ discount_rate = 0.0; discount_amount = 0.0; final_amount = 29999.0։

● Փոփոխություն՝ order_amount = 30000։ Սպասվող՝ discount_rate = 0.05; discount_amount = 1500.0; final_amount = 28500.0։

````python
# լուծումը
order_amount = 60000

if order_amount >= 60000:
    discount_rate = 0.10
elif order_amount >= 30000:
    discount_rate = 0.05
else:
    discount_rate = 0.0

discount_amount = round(order_amount * discount_rate, 2)
final_amount = round(order_amount - discount_amount, 2)
````

````python
assert discount_rate == 0.10, 'Նախ ստուգեք բարձր threshold-ը։'
assert discount_amount == 6000.0, 'Հաշվեք զեղչի գումարը։'
assert final_amount == 54000.0, 'Զեղչը հանեք original amount-ից։'
print('Առաջադրանք 3․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 4 Score validation**

**Տրված է**

score = 101

Score-ը number է։ Նախ validation արեք․ 0–100 միջակայքից դուրս score-ը invalid է։

Ստեղծեք grade, score_is_valid, passed․

● invalid → 'Invalid score', False, False,

● 90–100 → 'Excellent', True, True,

● 60-ից մինչև 90-ը չներառյալ → 'Passed', True, True,

● 0-ից մինչև 60-ը չներառյալ → 'Needs practice', True, False։

Invalid check-ը պետք է կատարվի classification-ից առաջ։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ score = 0։ Սպասվող՝ grade = 'Needs practice'; score_is_valid = True; passed = False։

● Փոփոխություն՝ score = 60։ Սպասվող՝ grade = 'Passed'; score_is_valid = True; passed = True։

````python
# լուծումը
score = 101

score_is_valid = 0 <= score <= 100

if not score_is_valid:
    grade = 'Invalid score'
    passed = False
elif score >= 90:
    grade = 'Excellent'
    passed = True
elif score >= 60:
    grade = 'Passed'
    passed = True
else:
    grade = 'Needs practice'
    passed = False
````

````python
assert grade == 'Invalid score', '101-ը վավեր score չէ։'
assert score_is_valid is False, 'Ստուգեք միջակայքի երկու սահմանները։'
assert passed is False, 'Invalid score-ը pass չի համարվում։'
print('Առաջադրանք 4․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 6 Country և delivery rule**

**Տրված է**

country = ' am '

basket_amount = 12000

is_member = True

Country-ն string է, amount-ը nonnegative number, member flag-ը՝ bool։ Ստեղծեք country_code՝ strip և uppercase արած country-ն։ Supported codes-ը AM, GE են։

Ստեղծեք delivery_supported, free_shipping, shipping_fee (AMD)․

● unsupported country → False, False, None,

● supported country-ում amount ≥15000 կամ (member է և amount ≥10000) → True, True, 0,

● supported մնացած orders → True, False, 1200։

Country check-ը պետք է գերակայի discount/membership rule-ին։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ country = 'fr'; basket_amount = 50000։ Սպասվող՝ country_code = 'FR'; delivery_supported = False; free_shipping = False; shipping_fee = None։

````python
# լուծումը
country = ' am '
basket_amount = 12000
is_member = True

country_code = country.strip().upper()

delivery_supported = country_code in ['AM', 'GE']

if not delivery_supported:
    free_shipping = False
    shipping_fee = None
elif basket_amount >= 15000 or (is_member and basket_amount >= 10000):
    free_shipping = True
    shipping_fee = 0
else:
    free_shipping = False
    shipping_fee = 1200
````

**Առաջադրանք 8 Conversion rate առանց zero division ի**

**Տրված է**

visits = 0

purchases = 0

Inputs-ը integers են։ Ստեղծեք rate_status, conversion_percent, has_rate․

1. Visits կամ purchases negative է, կամ purchases > visits → 'Invalid counts', None, False։
2. Վավեր counts-ի դեպքում visits=0 → 'No visits', None, False։
3. Մնացածը → 'Calculated', purchases / visits * 100՝ 2 decimal places-ով, True։

Սա ուսումնական metric է․ այս խնդրում purchases-ը չի կարող visits-ից մեծ լինել։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ visits = 3; purchases = 1։ Սպասվող՝ rate_status = 'Calculated'; conversion_percent = 33.33; has_rate = True։

● Փոփոխություն՝ visits = 10; purchases = 0։ Սպասվող՝ rate_status = 'Calculated'; conversion_percent = 0.0; has_rate = True։

**Առաջադրանք 9 Support ticket ի priority**

**Տրված է**

ticket = {'status': 'open', 'severity': 'medium', 'is_vip': True}

Dictionary-ի բոլոր keys-ը կան։ Status-ը 'open' կամ 'closed' է, severity-ն՝ 'low', 'medium', 'high' կամ 'critical', is_vip-ը՝ bool։

Ստեղծեք priority, response_hours, notify_manager՝ այս հերթականությամբ․

● Closed ticket → 'Closed', 0, False։

● Բաց critical ticket → 'Critical', 1, True։

● Բաց ticket, որը high է կամ VIP customer-ից է → 'High', 4, False։

● Մնացած բաց tickets → 'Normal', 24, False։

Closed case-ը պետք է գերակայի մյուս conditions-ին։ Original dictionary-ն մի փոփոխեք։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ ticket = {'status': 'closed', 'severity': 'critical', 'is_vip': True}։ Սպասվող՝ priority = 'Closed'; response_hours = 0; notify_manager = False։

● Փոփոխություն՝ ticket = {'status': 'open', 'severity': 'critical', 'is_vip': False}։ Սպասվող՝ priority = 'Critical'; response_hours = 1; notify_manager = True։

````python
ticket = {'status': 'open', 'severity': 'medium', 'is_vip': True}
# ticket = {'status': 'open', 'severity': 'critical', 'is_vip': False}

if ticket['status'] == 'closed':
    priority = 'Closed'
    response_hours = 0
    notify_manager = False
else:
    if ticket['severity'] == 'critical':
        priority = 'Critical'
        response_hours = 1
        notify_manager = True
    elif ticket['severity'] == 'high' or ticket['is_vip']:
        priority = 'High'
        response_hours = 4
        notify_manager = False
    else:
        priority = 'Normal'
        response_hours = 24
        notify_manager = False
````

````python
assert priority == 'High', 'VIP flag-ը կարող է բարձրացնել priority-ն։'
assert response_hours == 4, 'High ticket-ի response-ը 4 hours է։'
assert notify_manager is False, 'Manager-ը ծանուցվում է միայն բաց critical case-ում։'
print('Առաջադրանք 9․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 10 Անկախ quality flags**

email = ' '

● missing_email՝ strip արած email-ը դատարկ է,

Ստեղծեք flag_count՝ True flags-ի քանակը (sum()-ով), և review_status՝ 'Review', եթե որևէ flag True է, այլապես 'Clear'։ Status-ի համար օգտագործեք if / else։ Flags-ը անկախ են և կարող են միաժամանակ True լինել։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ email = 'ani@example.com'; purchase_count = 1; monthly_visits = 3։ Սպասվող՝ quality_flags = {'missing_email': False, 'no_purchases': False, 'low_activity': False}; flag_count = 0; review_status = 'Clear'։

raw_channel = ' Email '

● Փոփոխություն՝ raw_channel = 'WEB'։ Սպասվող՝ channel = 'web'; handler = 'Web team'; automatic_reply = True։

● Փոփոխություն՝ raw_channel = 'phone'։ Սպասվող՝ channel = 'phone'; handler = 'Call team'; automatic_reply = False։

````python
raw_channel = ' Email '
channel = raw_channel.strip().lower()

match channel:
    case 'web':
        handler = 'Web team'
        automatic_reply = True
    case 'email':
        handler = 'Inbox team'
        automatic_reply = True
    case 'phone':
        handler = 'Call team'
        automatic_reply = False
    case _:
        handler = 'Manual review'
        automatic_reply = False
````

````python
assert channel == 'email', 'Normalize արեք text-ը մինչև match-ը։'
assert handler == 'Inbox team', 'Ընտրեք email case-ը։'
assert automatic_reply is True, 'Email case-ը ունի automatic reply։'
print('Առաջադրանք 11․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 12 match / case OR patterns**

**Տրված է**

raw_format = ' CSV '

Ստեղծեք normalized_format՝ strip և lowercase արած input-ը։ match / case-ում օգտագործեք OR patterns՝ |-ով․

● 'csv' կամ 'tsv' → format_group = 'Tabular',

● 'json' կամ 'yaml' → 'Structured',

● 'txt' կամ 'md' → 'Text',

● մյուս values-ը → 'Unsupported'։

format_supported-ը True է առաջին երեք խմբերի դեպքում, այլապես False։ Մի օգտագործեք boolean or keyword-ը pattern-ի ներսում։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ raw_format = 'TSV'։ Սպասվող՝ normalized_format = 'tsv'; format_group = 'Tabular'; format_supported = True։

● Փոփոխություն՝ raw_format = 'yaml'։ Սպասվող՝ format_group = 'Structured'; format_supported = True։

````python
raw_format = ' CSV '
normalized_format = raw_format.strip().lower()

match normalized_format:
    case 'csv' | 'tsv':
        format_group = 'Tabular'
        format_supported = True
    case 'json' | 'yaml':
        format_group = 'Structured'
        format_supported = True
    case 'txt' | 'md':
        format_group = 'Text'
        format_supported = True
    case _:
        format_group = 'Unsupported'
        format_supported = False
````

````python
assert normalized_format == 'csv', 'Input-ը normalize արեք։'
assert format_group == 'Tabular', 'CSV-ն tabular խմբում է։'
assert format_supported is True, 'Supported format-ի flag-ը True է։'
print('Առաջադրանք 12․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 13 match / case Inventory guards**

**Տրված է**

available_units = 3

Input-ը integer կամ None է։ Օգտագործեք match / case և guards (case value if ...)։ Ստեղծեք inventory_status, order_allowed, inventory_action․

● None → 'Missing', False, 'Review data',

● negative → 'Invalid', False, 'Review data',

● 0 → 'Empty', False, 'Restock',

● 1–4 → 'Limited', True, 'Restock soon',

● 5 և ավելի → 'Available', True, 'No action'։

Guard-երի հերթականությունը կարևոր է։ None-ը ստուգեք numeric comparisons-ից առաջ։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ available_units = None։ Սպասվող՝ inventory_status = 'Missing'; order_allowed = False; inventory_action = 'Review data'։

● Փոփոխություն՝ available_units = -1։ Սպասվող՝ inventory_status = 'Invalid'; order_allowed = False; inventory_action = 'Review data'։

````python
available_units = 3

match available_units:
    case None:
        inventory_status = 'Missing'
        order_allowed = False
        inventory_action = 'Review data'
    case n if n < 0:
        inventory_status = 'Invalid'
        order_allowed = False
        inventory_action = 'Review data'
    case 0:
        inventory_status = 'Empty'
        order_allowed = False
        inventory_action = 'Restock'
    case n if 1 <= n <= 4:
        inventory_status = 'Limited'
        order_allowed = True
        inventory_action = 'Restock soon'
    case n if n >= 5:
        inventory_status = 'Available'
        order_allowed = True
        inventory_action = 'No action'
````

````python
assert inventory_status == 'Limited', '3-ը Limited միջակայքում է։'
assert order_allowed is True, 'Դրական stock-ը թույլ է տալիս order։'
assert inventory_action == 'Restock soon', 'Ընտրեք համապատասխան action-ը։'
print('Առաջադրանք 13․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 14 match / case Transaction tuple**

**Տրված է**

event = ('refund', 12000)

Event-ը երկու-element tuple կամ list է՝ (kind, amount)։ Kind-ը string է, amount-ը nonnegative number։

Օգտագործեք sequence patterns և capture variable․

● ('sale', amount) → event_label = 'Sale', signed_amount = amount,

● ('refund', amount) → 'Refund', signed_amount = -amount,

● մյուս cases → 'Unknown', signed_amount = None։

recognized_event-ը True է sale/refund case-երում, այլապես False։ Սա signed amount-ի ուսումնական գրանցում է, ոչ շահույթի հաշվարկ։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ event = ('sale', 12000)։ Սպասվող՝ event_label = 'Sale'; signed_amount = 12000; recognized_event = True։

● Փոփոխություն՝ event = ['refund', 0]։ Սպասվող՝ event_label = 'Refund'; signed_amount = 0; recognized_event = True։

````python
event = ('refund', 12000)

match event:
    case ('sale', amount):
        event_label = 'Sale'
        signed_amount = amount
        recognized_event = True
    case ('refund', amount):
        event_label = 'Refund' 
        signed_amount = -amount
        recognized_event = True
    case _:
        event_label = 'Unknown'
        signed_amount = None
        recognized_event = False
````

````python
assert event_label == 'Refund', 'Pattern-ում kind-ը գրեք quotes-ով։'
assert signed_amount == -12000, 'Refund-ի amount-ը negative գրանցեք։'
assert recognized_event is True, 'Refund-ը ճանաչված event է։'
print('Առաջադրանք 14․ հիմնական ստուգումներն անցան։')
````

**Առաջադրանք 15 Mini project Order decision report**

**Տրված է**

```
order = {
    'name': ' Mari ',
    'amount': 45000,
    'is_active': True,
    'is_member': True,
    'payment_status': ' PAID ',
}
```
Կառուցեք ամբողջական decision report՝ օգտագործելով if / elif / else և match / case։ Բոլոր keys-ը կան։

Name/payment_status-ը strings են, amount-ը՝ finite number կամ None, flags-ը՝ booleans։ Original dictionary-ն մի փոփոխեք։

**Քայլ 1**․ Normalize. Name-ը՝ strip և title, payment_status-ը՝ strip և lowercase։

**Քայլ 2**․ Status-ի priority. Առաջին համապատասխան rule-ն է ընտրվում․

1. Cleaned name-ը դատարկ է → 'Invalid name'։
2. Amount-ը None է → 'Missing amount'։
3. Amount-ը negative է → 'Invalid amount'։
4. Customer-ը inactive է → 'Blocked'։
5. Payment-ը 'paid' է → 'Ready', 'pending' → 'Awaiting payment', 'cancelled' → 'Cancelled', մնացածը → 'Unknown payment'։

**Քայլ 3**․ Հաշվարկ. Միայն Ready case-ի համար ընտրեք discount rate-ը․ member է և amount ≥40000 → 0.10, այլապես amount ≥25000 → 0.05, մնացածը → 0.0։ Payable-ը amount * (1 - rate) է՝ 2 decimal places-ով։ Non-ready cases-ում rate-ը և payable-ը None են։

**Քայլ 4**․ Next action. match / case-ով status-ը կապեք action-ին․

● Ready → 'Process order',

● Awaiting payment → 'Wait for payment',

● Cancelled → 'Close order',

● Blocked → 'Contact support',

● Unknown payment → 'Review payment',

● մնացած statuses → 'Correct data'։ ## էստեղ անհապատասխանություն կա

Արդյունք. order_report dictionary՝ ճիշտ 6 keys-ով՝ name, status, discount_rate, payable_amd, next_action, currency։ Currency-ն միշտ 'AMD' է։

Comment-ով բացատրեք՝ ինչու non-ready case-ում payable-ը None է, ոչ 0։

**Լրացուցիչ ինքնաստուգում**

● Փոփոխություն՝ order = {'name': ' Mari ', 'amount': 45000, 'is_active': True, 'is_member': True, 'payment_status': 'pending'}։ Սպասվող՝ order_report = {'name': 'Mari', 'status': 'Awaiting payment', 'discount_rate': None, 'payable_amd': None, 'next_action': 'Wait for payment', 'currency': 'AMD'}։

● Փոփոխություն՝ order = {'name': ' Mari ', 'amount': 39999, 'is_active': True, 'is_member': True, 'payment_status': ' PAID '}։ Սպասվող՝ order_report = {'name': 'Mari', 'status': 'Ready', 'discount_rate': 0.05, 'payable_amd': 37999.05, 'next_action': 'Process order', 'currency': 'AMD'}։

````python
order = {
    'name': ' Mari ',
    'amount': 45000,
    'is_active': True,
    'is_member': True,
    'payment_status': ' PAID ',
}

normalized_name = order['name'].strip().title()
normalized_payment_status = order['payment_status'].strip().lower()

order_report = {
    'name': normalized_name,
    'status': normalized_payment_status,
    'currency': 'AMD'
}

if normalized_name == '':
    order_report['status'] = 'Invalid name'
elif order['amount'] is None:
    order_report['status'] = 'Missing amount'
elif order['amount'] < 0:
    order_report['status'] = 'Invalid amount'
elif not order['is_active']:
    order_report['status'] = 'Blocked'
else: 
    match order_report['status']:
        case 'paid':
            order_report['status'] = 'Ready'
            order_report['next_action'] = 'Process order'
        case 'pending':
            order_report['status'] = 'Awaiting payment'
            order_report['next_action'] = 'Wait for payment'
        case 'cancelled':
            order_report['status'] = 'Cancelled'
            order_report['next_action'] = 'Close order'
        case _:
            order_report['status'] = 'Unknown payment'
            order_report['next_action'] = 'Review payment'

if order_report['status'] == 'Ready':
    if order['is_member'] == True and order['amount'] >= 40000:
        order_report['discount_rate'] = 0.1
    elif order['amount'] >= 25000:
        order_report['discount_rate'] = 0.05
    else:
        order_report['discount_rate'] = 0.0
    
    order_report['payable_amd'] = round(order['amount'] * (1 - order_report['discount_rate']), 2)
else:
    order_report['discount_rate'] = None
    order_report['payable_amd'] = None
````

````python
assert (
set[str](order_report) == {
'name', 'status', 'discount_rate',
'payable_amd', 'next_action', 'currency'
}
), 'Report-ը պետք է ունենա պահանջված 6 keys-ը։'
assert (
order_report['name'] == 'Mari'
and order_report['status'] == 'Ready'
), 'Ստուգեք normalization-ը և status-ը։'
assert (
order_report['discount_rate'] == 0.1
and order_report['payable_amd'] == 40500.0
), 'Կիրառեք member discount-ը։'
assert (
order_report['next_action'] == 'Process order'
and order_report['currency'] == 'AMD'
), 'Match-ով ընտրեք ճիշտ action-ը։'
print('Առաջադրանք 15․ հիմնական ստուգումներն անցան։')
````
