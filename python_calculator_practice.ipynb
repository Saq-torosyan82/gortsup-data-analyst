**Inputs and Types**

```python
product_name = "Bag"
#quantity = 4
quantity = 1
unit_price = 2500
unit_cost = 1500
discount_percentage = 10
is_return = False
print(f"type of quantity: {type(quantity)}")
print(f"type of discount_percentage: {type(discount_percentage)}")
print(f"type of product_name: {type(product_name)}")
print(f"type of is_return: {type(is_return)}")
```

**Calculations**

```python
revenue_before_discount = quantity * unit_price
discount_amount = revenue_before_discount * discount_percentage / 100
sales_revenue = revenue_before_discount - discount_amount
total_product_cost = quantity * unit_cost
gross_profit = sales_revenue - total_product_cost

print(f"Revenue before discount: {revenue_before_discount} AMD")
print(f"Discount amount: {discount_amount} AMD")
print(f"Sales revenue: {sales_revenue} AMD")
print(f"Total product cost: {total_product_cost} AMD")
print(f"Gross profit: {gross_profit} AMD")
```

**Boolean Flag**

```python
is_successful_sale = sales_revenue >= 8000 and gross_profit > 0 and not is_return
print(f"Successful sale: {is_successful_sale}")
```
