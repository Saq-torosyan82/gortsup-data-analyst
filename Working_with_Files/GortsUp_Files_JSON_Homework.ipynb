## Առաջադրանք 1 · Տեքստային ֆայլի ստեղծում

Գրեք `write_note(path, text)` ֆունկցիան։ `"w"` ռեժիմով ստեղծեք կամ վերագրեք ֆայլը և գրեք `text` տողը՝ առանց լրացուցիչ նիշ ավելացնելու։ Վերադարձվող արժեք չի պահանջվում։


### Սկզբնական կառուցվածք


```python
def write_note(path, text):
    with open(path, "w", encoding="utf-8") as file:
        file.write(text)
```

### Ստուգումներ


```python
path = work_dir / "note.txt"
write_note(path, "Բարև, Python։")
with open(path, "r", encoding="utf-8") as file:
    assert file.read() == "Բարև, Python։"
write_note(path, "Նոր տեքստ")
with open(path, "r", encoding="utf-8") as file:
    assert file.read() == "Նոր տեքստ" 

print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
write_note(path, "New Line")
with open(path, "r", encoding="utf-8") as file:
    assert file.read() == "New Line"
print('passed my')
```

## Առաջադրանք 2 · Ֆայլի ամբողջական ընթերցում

Գրեք `read_note(path)` ֆունկցիան։ Կարդացեք գոյություն ունեցող տեքստային ֆայլի ամբողջ բովանդակությունը և վերադարձրեք որպես տող։ Պահպանեք բացատներն ու նոր տողերը։

## Առաջադրանք 3 · Նոր տողի ավելացում

Գրեք `append_line(path, text)` ֆունկցիան։ `"a"` ռեժիմով ավելացրեք `text`-ը և մեկ `\n` նիշ։ Եթե ֆայլը չկա, այն պետք է ստեղծվի։


### Սկզբնական կառուցվածք


```python
def append_line(path, text):
    with open(path, "a", encoding="utf-8") as file:
        file.write(text + "\n")
```

## Առաջադրանք 4 · Ստեղծել առանց վերագրելու

Գրեք `create_once(path, text)` ֆունկցիան։ Օգտագործեք `"x"` ռեժիմը։ Հաջող ստեղծման դեպքում վերադարձրեք `True`, `FileExistsError`-ի դեպքում՝ `False`։


### Սկզբնական կառուցվածք


```python
def create_once(path, text):
    try:
        with open(path, "x", encoding="utf-8") as file:
            file.write(text)
            return True
    except FileExistsError:
        return False
```

### Ստուգումներ


```python
path = work_dir / "only_once.txt"
path.unlink(missing_ok=True)
assert create_once(path, "Առաջին") is True
assert create_once(path, "Երկրորդ") is False
assert read_note(path) == "Առաջին" 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
path_my = work_dir / "only_once_my.txt"
path_my.unlink(missing_ok=True)
assert create_once(path_my, "Առաջին") is True
assert create_once(path_my, "Երկրորդ") is False
assert read_note(path_my) == "Առաջին" 
print('passed my')
```

## Առաջադրանք 5 · Ֆայլի ավտոմատ փակման ստուգում

Գրեք `check_closed(path)` ֆունկցիան։ `with open(..., "w", ...)` բլոկում գրեք `"test"` և պահեք `file.closed`-ի արժեքը։ Բլոկից հետո նորից ստուգեք `file.closed`-ը։ Վերադարձրեք `(inside, outside)`։


### Սկզբնական կառուցվածք


```python
def check_closed(path):
    inside = False
    outside = False
    with open(path, "w", encoding="utf-8") as file:
        file.write("test")
        inside = file.closed

    outside = file.closed
    return (inside, outside)
```

### Ստուգումներ


```python
path = work_dir / "closed.txt"
assert check_closed(path) == (False, True)
assert read_note(path) == "test" 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
path_my = work_dir / "closed_my.txt"
assert check_closed(path_my) == (False, True)
assert read_note(path_my) == "test" 
print('passed my')
```

## Առաջադրանք 6 · Բացակայող ֆայլի անվտանգ ընթերցում

Գրեք `safe_read(path)` ֆունկցիան։ Վերադարձրեք ֆայլի բովանդակությունը։ Միայն `FileNotFoundError`-ի դեպքում վերադարձրեք `None`։


### Սկզբնական կառուցվածք


```python
def safe_read(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            return file.read()
    except FileNotFoundError:
        return None
```

### Ստուգումներ


```python
missing = work_dir / "missing.txt"
missing.unlink(missing_ok=True)
assert safe_read(missing) is None
path = work_dir / "empty.txt"
write_note(path, "")
assert safe_read(path) == ""
write_note(path, "տվյալներ")
assert safe_read(path) == "տվյալներ" 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
missing_my = work_dir / "missing_my.txt"
missing.unlink(missing_ok=True)
assert safe_read(missing_my) is None
path_my = work_dir / "empty_my.txt"
write_note(path_my, "")
assert safe_read(path_my) == ""
write_note(path_my, "տվյալներ")
assert safe_read(path_my) == "տվյալներ"
print('passed my')
```

## Առաջադրանք 7 · Ոչ դատարկ տողերի ընթերցում

Գրեք `read_nonempty_lines(path)` ֆունկցիան։ Յուրաքանչյուր տողի եզրային բացատներն ու նոր տողի նիշը հեռացրեք `strip()`-ով, դատարկ արդյունքները բաց թողեք։


### Սկզբնական կառուցվածք


```python
def read_nonempty_lines(path):
    with open(path, "r", encoding="utf-8") as file:
        return [line.strip() for line in file.readlines() if line.strip()]
```

### Ստուգումներ


```python
path = work_dir / "names.txt"
write_note(path, " Անի \n\n \nԱրամ\n")
assert read_nonempty_lines(path) == ["Անի", "Արամ"]
write_note(path, " \n")
assert read_nonempty_lines(path) == [] 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
path_my = work_dir / "names_my.txt"
write_note(path_my, " John \n\n \nDoe\n")
assert read_nonempty_lines(path_my) == ["John", "Doe"]
print('passed my')
```

## Առաջադրանք 8 · Տեքստային թվերի մշակում

Գրեք `summarize_numbers(path)` ֆունկցիան։ Ոչ դատարկ տողերը փորձեք փոխարկել `int()`-ով։ `ValueError`-ի դեպքում ավելացրեք `invalid` հաշվիչը։


### Սկզբնական կառուցվածք


```python
def summarize_numbers(path):
    count = 0
    total = 0
    invalid = 0
    with open(path, "r", encoding="utf-8") as file:
        for line in file.readlines():
            if line.strip():
                try:
                    total += int(line.strip())
                    count += 1
                except ValueError:
                    invalid += 1

    return {"count": count, "total": total, "invalid": invalid}
```

### Ստուգումներ


```python
path = work_dir / "numbers.txt"
write_note(path, "10\n-2\nhello\n\n3.5\n0\n")
assert summarize_numbers(path) == {"count": 3, "total": 8, "invalid": 2}
write_note(path, "")
assert summarize_numbers(path) == {"count": 0, "total": 0, "invalid": 0} 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
path_my = work_dir / "numbers_my.txt"
write_note(path_my, "5\n-12\nhello\n\n5.5\n0\n")
assert summarize_numbers(path_my) == {"count": 3, "total": -7, "invalid": 2}
print('passed my')
```

## Առաջադրանք 9 · CSV ֆայլի ընթերցում

Գրեք `read_products(path)` ֆունկցիան՝ `csv.DictReader`-ով։ Վերադարձրեք տողերի ցուցակը՝ յուրաքանչյուր տողը բառարան։ Թվային տեսք ունեցող արժեքները պահեք տողերի տեսքով։


### Տրված տվյալներ


```python
path = work_dir / "products_input.csv"
write_note(path, "product,quantity,price\nԳրիչ,2,150\nՏետր,3,400\n")
```

### Սկզբնական կառուցվածք


```python
def read_products(path):
    with open(path, "r", encoding="utf-8") as file:
        csv_reader = csv.DictReader(file)
        return list(csv_reader)
```

### Ստուգումներ


```python
rows = read_products(path)
assert len(rows) == 2
assert rows[0] == {"product": "Գրիչ", "quantity": "2", "price": "150"}
write_note(path, "product,quantity,price\n")
assert read_products(path) == [] 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
path_my = work_dir / "products_input_my.csv"
write_note(path_my, "product,quantity,price\nBook,3,300\nPen,5,500\n")
rows_my = read_products(path_my)
assert len(rows_my) == 2
assert rows_my[0] == {"product": "Book", "quantity": "3", "price": "300"}
print('passed my')
```

## Առաջադրանք 10 · CSV ֆայլի գրառում

Գրեք `write_products(path, rows)` ֆունկցիան՝ `csv.DictWriter`-ով։ Սյունակների հերթականությունը՝ `product, quantity, price`։ Գրեք վերնագիրն ու բոլոր տողերը։


### Տրված տվյալներ


```python
products = [{"product": "Գրիչ, կապույտ", "quantity": 2, "price": 150}]
```

### Սկզբնական կառուցվածք


```python
def write_products(path, rows):
    with open(path, "w", encoding="utf-8") as file:
        csv_writer = csv.DictWriter(file, ["product", "quantity", "price"])
        csv_writer.writeheader()
        csv_writer.writerows(rows)
```

### Ստուգումներ


```python
path = work_dir / "products_output.csv"
write_products(path, products)
assert read_products(path) == [{"product": "Գրիչ, կապույտ", "quantity": "2", "price": "150"}]
write_products(path, [])
with open(path, "r", encoding="utf-8", newline="") as file:
    assert list(csv.reader(file)) == [["product", "quantity", "price"]] 
print('passed')
```

## Առաջադրանք 12 · JSON ֆայլի անվտանգ բեռնում

Գրեք `load_json_safe(path)` ֆունկցիան։ Հաջողության դեպքում վերադարձրեք `("ok", data)`, `FileNotFoundError`-ի դեպքում՝ `("missing", None)`, `json.JSONDecodeError`-ի դեպքում՝ `("invalid", None)`։


### Սկզբնական կառուցվածք


```python
def load_json_safe(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            return ("ok", json.load(file))
    except FileNotFoundError:
        return ("missing", None)
    except json.JSONDecodeError:
        return ("invalid", None)
```

### Ստուգումներ


```python
path = work_dir / "json_check.json"
write_note(path, '{"count": 2}')
assert load_json_safe(path) == ("ok", {"count": 2})
write_note(path, '{"count": 2,}')
assert load_json_safe(path) == ("invalid", None)
path.unlink()
assert load_json_safe(path) == ("missing", None)
write_note(path, "null")
assert load_json_safe(path) == ("ok", None) 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
path = work_dir / "json_check_my.json"
write_note(path, '{"check": "ok"}')
assert load_json_safe(path) == ("ok", {"check": "ok"})
print('passed my')
```

## Առաջադրանք 14 · CSV տվյալների մաքրում

Գրեք `clean_product_rows(rows)` ֆունկցիան։ Մաքրեք `product`, փոխարկեք `quantity`-ն `int()`-ով, `price`-ը՝ `float()`-ով։ Ընդունեք միայն ոչ դատարկ ապրանք, դրական քանակ և դրական գին։ Վերադարձրեք `(cleaned, rejected_count)`։


### Տրված տվյալներ


```python
rows = [
    {"product": " Գրիչ ", "quantity": "2", "price": "150"},
    {"product": "", "quantity": "1", "price": "100"},
    {"product": "Տետր", "quantity": "two", "price": "400"},
    {"product": "Գիրք", "quantity": "1", "price": "0"}
]
```

### Սկզբնական կառուցվածք


```python
def clean_product_rows(rows):
    cleaned = []
    rejected = 0
    for row in rows:
        product = row["product"].strip()
        if product:
            try:
                quantity = int(row["quantity"])
                price = float(row["price"])
                if quantity > 0 and price > 0:
                    cleaned.append({
                        "product": product,
                        "quantity": quantity,
                        "price": price,
                        "revenue": quantity * price
                    })
                else:
                    rejected += 1
            except ValueError:
                rejected += 1
        else:
            rejected += 1

    return cleaned, rejected
```

### Ստուգումներ


```python
original = [dict(row) for row in rows]
cleaned, rejected = clean_product_rows(rows)
assert cleaned == [{"product": "Գրիչ", "quantity": 2, "price": 150.0, "revenue": 300.0}]
assert rejected == 3
assert rows == original
assert clean_product_rows([]) == ([], 0) 
print('passed')
```

### Սեփական ստուգում
Լուծումից հետո ավելացրեք առնվազն մեկ սեփական օրինակ և բացատրեք ակնկալվող արդյունքը։


```python
# Գրեք այստեղ ձեր սեփական ստուգումը
rows = [
    {"product": " Book ", "quantity": "2", "price": "150"},
    {"product": "", "quantity": "1", "price": "100"},
    {"product": "Pen", "quantity": "fff", "price": "500"},
    {"product": "Book", "quantity": "-1", "price": "500"}
]

cleaned, rejected = clean_product_rows(rows)
assert cleaned == [{"product": "Book", "quantity": 2, "price": 150.0, "revenue": 300.0}]
assert rejected == 3 
print('passed my')
```

## Առաջադրանք 15 · Ամբողջական ֆայլային հաշվետվություն

Գրեք `build_file_report(source_path, json_path, log_path)` ֆունկցիան։ Օգտագործեք `read_products()` և `clean_product_rows()`։ Կազմեք JSON հաշվետվություն և երեք տողանոց log։


### Տրված տվյալներ


```python
source = work_dir / "sales.csv"
target = work_dir / "sales_report.json"
log = work_dir / "sales_log.txt"
write_note(source, "product,quantity,price\nԳրիչ,2,150\nՏետր,2,400\nԳիրք,0,500\n")
```

### Սկզբնական կառուցվածք


```python
def build_file_report(source_path, json_path, log_path):
    rows = read_products(source_path)
    cleaned, rejected = clean_product_rows(rows)
    report = {
        "input_count": len(rows),
        "accepted_count": len(cleaned),
        "rejected_count": rejected,
        "total_revenue": sum(row["revenue"] for row in cleaned),
        "records": cleaned
    }
    with open(json_path, "w", encoding="utf-8") as file:
        json.dump(report, file, ensure_ascii=False, indent=2)
    with open(log_path, "w", encoding="utf-8") as file:
        file.write(f"input={len(rows)}\naccepted={len(cleaned)}\nrejected={rejected}\n")
    return report
```

### Ստուգումներ


```python
report = build_file_report(source, target, log)
assert report == {
    "input_count": 3,
    "accepted_count": 2,
    "rejected_count": 1,
    "total_revenue": 1100.0,
    "records": [
        {"product": "Գրիչ", "quantity": 2, "price": 150.0, "revenue": 300.0},
        {"product": "Տետր", "quantity": 2, "price": 400.0, "revenue": 800.0}
    ]
}
with open(target, "r", encoding="utf-8") as file:
    assert json.load(file) == report
assert read_note(log) == "input=3\naccepted=2\nrejected=1\n" 
print('passed')
```
